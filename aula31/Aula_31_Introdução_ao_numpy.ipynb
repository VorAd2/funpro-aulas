## 1. Do tipo `list` ao `ndarray`

Nas aulas anteriores, usamos listas para guardar sequências de dados. Em Python, uma lista aceita tipos variados (inclusive objetos mais complexos, como dicionários!).

A biblioteca NumPy (Numerical Python) introduz o tipo `ndarray` (N-dimensional array):
- Homogêneo: todos os elementos têm obrigatoriamente o mesmo tipo.
- Memória contígua: dados ficam alinhados sequencialmente na memória RAM, o que acelera o acesso.

Com essas regras, acessar e calcular elementos fica mais eficiente!

Para acessar uma biblioteca, utilizamos o comando `import <nome da biblioteca>`. Podemos usar a cláusula `as <apelido>` para dar um apelido.

```python
import numpy as np
# a partir de agora, acessamos as funções da biblioteca a partir da variável np
# np.funcao1, etc.
print(np)
print(type(np))
```

Array/vetor é o nome que se dá à estrutura de dados que agrupa de modo contíguo dados na memória.

```python
lista_base = [10, 20, 30, 40]
vetor = np.array(lista_base)
```

```python
print(vetor)
```

## 2. Operações vetorizadas versus laços for

Até agora, para calcular média ou variância, precisávamos percorrer a lista usando um laço `for`.

No NumPy, as operações são vetorizadas: a operação aritmética é aplicada a cada elemento do array diretamente em baixo nível (linguagem C), sem a necessidade de um laço explícito em Python.

### Exemplo: notas de uma turma

Vamos representar as notas de 5 estudantes em uma avaliação.

Relembrando como fazíamos para aplicar um ponto extra em cada elemento de uma lista.

```python
# ABORDAGEM TRADICIONAL
notas_lista = [0.5, 7.0, 3.0, 8.5, 2.5]
notas_ajustadas = []

for nota in notas_lista:
    notas_ajustadas.append(nota + 1.0)

print(notas_ajustadas)
```

De modo vetorizado, fica bem mais conciso!

```python
notas = np.array([0.5, 7.0, 3.0, 8.5, 2.5])
print(notas)
```

```python
# ABORDAGEM VETORIZADA (NumPy)
notas_ajustadas_np = notas + 1.0
print(notas_ajustadas_np)
```

A operação aritmética `+ 1.0` é propagada para cada posição do array.

Isso funciona para muitas das operações com escalar:

```python
print(notas - 2)
print(notas * 2)
print(notas / 2)
print(notas ** 2)
```

Também funciona entre arrays de tamanhos compatíveis:

```python
print(np.array([100, 101, 102]) + np.array([1, 2, 3]))
print(np.array([100, 101, 102]) - np.array([1, 2, 3]))
print(np.array([100, 101, 102]) * np.array([1, 2, 3]))
print(np.array([100, 101, 102]) / np.array([1, 2, 3]))
print(np.array([100, 101, 102]) ** np.array([1, 2, 3]))
```

O que significa tamanho ser compatível? Vamos testar:

```python
print(np.array([100, 101, 102]) + np.array([1, 2])) # tamanho incompatível
```

```python
print(np.array([100, 101, 102]) + np.array([4])) # broadcast: escalar 4 é interpretado como [4, 4, 4]
```

```python
print(np.array([100, 101, 102]) + np.array([1, 2, 3, 4, 5, 6])) # tamanho incompatível
```

**Regra de broadcast:** tamanhos são compatíveis ao longo de uma dimensão quando
1. tem o mesmo tamanho, ou
2. um dos tamanhos é 1, caso no qual esse tamanho é "esticado" para se igualar ao outro

O que significa esticar?

```python
arr = np.array([4])
print(arr)
print(arr.shape)
```

```python
arr_esticado = np.broadcast_to(arr, (3,)) # isso é feito implicitamente no broadcast
print(arr_esticado)
print(arr_esticado.shape)
```

No momento, estamos trabalhando com estruturas 1D (1 dimensão; listas de números), então ao fazer uma operação entre arrays de shape $(N,)$ e $(M,)$ basta checar se $N = M$ ou $N = 1$ ou $M = 1$.

## 3. Indexação, fatiamento e filtragem

Já sabemos acessar e fatiar listas com colchetes (`lista[i]`). O NumPy mantém essa sintaxe, mas adiciona a filtragem booleana (máscaras), eliminando blocos `if` para seleção de subconjuntos de dados.

### Exemplo: tempos de resposta em um experimento

Considere uma amostra com tempos de reação (em segundos) registrados em um teste.

```python
tempos = np.array([0.42, 0.55, 0.38, 0.91, 0.45, 0.39, 1.20, 0.48])
print(tempos)
```

### Indexação posicional

Funciona de forma idêntica às listas (índice base zero e índices negativos para o final).

```python
primeiro = tempos[0]
print("Primeiro elemento:", primeiro)
```

```python
ultimo = tempos[-1]
print("Último elemento:", ultimo)
```

### Fatiamento (slicing)

Extração de um intervalo contíguo usando a sintaxe `inicio:fim`.

```python
tres_primeiros = tempos[0:3]
print(tres_primeiros)
```

### Atenção! Visualização (view) versus cópia

Em listas, um fatiamento cria uma cópia nova. No NumPy, o fatiamento cria apenas uma "janela" (*view*) sobre o mesmo espaço de memória. Alterar a fatia altera o array original e vice-versa!

```python
tempos = np.array([0.42, 0.55, 0.38, 0.91, 0.45, 0.39, 1.20, 0.48])
fatia = tempos[0:2]
fatia[0] = 9.99
```

```python
# O array original foi modificado
print(tempos)
```

```python
# Para criar uma estrutura independente, use o método copy()
tempos_copia = tempos.copy()
```

```python
tempos[0] = 0.42
```

```python
print(fatia)
```

```python
print(tempos_copia)
```

### Filtragem booleana (máscaras)

Em Python, filtrar exigia um `for` com `if` interno para dar `append` em uma lista nova. No NumPy, passamos um vetor de booleanos dentro dos colchetes.

Definindo um critério estatístico: identificar observações lentas (acima da média da amostra).

```python
limiar = tempos.mean()
print("Média da amostra:", limiar)
```

```python
mascara = tempos > limiar
print("Vetor de booleanos (máscara):", mascara)
```

```python
lentos = tempos[mascara]
print("Apenas valores acima da média:", lentos)
```

### Combinando operadores lógicos em filtros

Para combinar condições lógicas elemento a elemento, usamos `&` (and) e `|` (or) com cada condição entre parênteses.

```python
A = np.array([False, False, True, True])
B = np.array([False, True, False, True])
print(A & B)
print(A | B)
```

```python
# Filtrar tempos dentro do intervalo considerado normal: entre 0.40 e 0.60
condicao_1 = tempos >= 0.40
condicao_2 = tempos <= 0.60
condicao_final = condicao_1 & condicao_2
dentro_do_padrao = tempos[condicao_final]

print(tempos)
print('condicao_1:', condicao_1)
print('condicao_2:', condicao_2)
print('condicao_final:', condicao_final)
print(dentro_do_padrao)
```

## 4. Arrays multidimensionais (matrizes 2D)

Em aulas anteriores, vimos que matrizes são representadas no Python (sem bibliotecas) como listas aninhadas (acessando elementos com `lista[i][j]`). No NumPy, a sintaxe recomendada `matriz[i, j]` (a sintaxe tradicional funciona, mas é mais lenta).

### Exemplo: matriz de observações

Linhas representam 4 indivíduos da amostra. Colunas representam 3 medições (ex: AP1, AP2 e AP3).

```python
dados = [
    [7.0, 8.5, 6.0],
    [4.0, 5.0, 6.5],
    [9.0, 9.5, 8.0],
    [6.0, 6.5, 7.0]
]
matriz_notas = np.array(dados)
```

```python
print(matriz_notas)
print(matriz_notas.dtype)
```

### Inspecionando as dimensões

O atributo `shape` retorna uma tupla com a quantidade de linhas e colunas: `(linhas, colunas)`.

```python
print("Formato (linhas, colunas):", matriz_notas.shape)
```

```python
print("Dimensões:", matriz_notas.ndim)
```

```python
print("len:", len(matriz_notas))
```

```python
print("size:", matriz_notas.size)
```

### Indexação bidimensional

Podemos usar tanto `array[i][j]` quanto `array[linha, coluna]`.

```python
# Nota do primeiro estudante (linha 0) na segunda avaliação (coluna 1)
nota_especifica = matriz_notas[0, 1]
print("Nota pontual:", nota_especifica)
```

```python
print(matriz_notas[0][1]) # também podemos usar a indexação aninhada
```

Só que `array[linha, coluna]` é mais eficiente internamente, pois não é necessário criar o objeto temporário `L = array[linha]` para só então acessar `L[coluna]`.

### Fatiamento por linhas e colunas

O operador `:` isolado significa "selecionar todos os elementos deste eixo".

```python
# Todas as notas da primeira avaliação (todas as linhas da coluna 0)
prova_1 = matriz_notas[:, 0]
print("Notas da AP1:", prova_1)
```

```python
# Todas as notas do último estudante (linha 3, todas as colunas)
aluno_final = matriz_notas[3, :]
print("Notas do último estudante:", aluno_final)
```

### Agregação por eixo (axis)

Em Python (sem bibliotecas), calcular a média de cada prova exigiria laços aninhados. No NumPy, controlamos a direção da operação pelo parâmetro `axis`:
* `axis=0`: colapsa as linhas (opera verticalmente ao longo das colunas).
* `axis=1`: colapsa as colunas (opera horizontalmente ao longo das linhas).

Média global (todos os valores da matriz):

```python
media_geral = matriz_notas.mean()
print("Média global da turma:", media_geral)
```

Média por coluna (média de cada prova):

```python
media_provas = matriz_notas.mean(axis=0)
print("Média de cada prova:", media_provas)
print("shape:", media_provas.shape)
```

```python
media_provas = matriz_notas.mean(axis=0, keepdims=True)
print("Média de cada prova (keepdims=True):", media_provas)
print("shape:", media_provas.shape)
```

Média por linha (média final de cada estudante):

```python
media_estudantes = matriz_notas.mean(axis=1)
print("Média de cada estudante:", media_estudantes)
print("shape:", media_estudantes.shape)
```

```python
media_estudantes = matriz_notas.mean(axis=1, keepdims=True)
print("Média de cada estudante (keepdims=True):", media_estudantes)
print("shape:", media_estudantes.shape)
```

Desvio padrão de cada prova (dispersão das notas coluna a coluna):

```python
desvio_provas = matriz_notas.std(axis=0)
print("Desvio padrão por prova:", desvio_provas)
print("shape:", desvio_provas.shape)
```

```python
desvio_provas = matriz_notas.std(axis=0, keepdims=True)
print("Desvio padrão por prova (keepdims=True):", desvio_provas)
print("shape:", desvio_provas.shape)
```

Sabemos usar a função `range()` para gerar sequências de números inteiros para usar em laços de repetição. O NumPy expande essa ideia para valores de ponto flutuante e adiciona geradores para simulações estatísticas.

```python
print(grid)
```

### Reprodutibilidade científica e sementes (seed)

Computadores não geram números verdadeiramente aleatórios; usam algoritmos matemáticos determinísticos (pseudoaleatórios).

Fixar a semente (`seed`) garante que o mesmo experimento ou simulação gere exatamente os mesmos resultados toda vez que o código for executado.

```python
# Criando um gerador de números pseudoaleatórios
rng = np.random.default_rng(seed=42)
```

```python
print(rng)
```

É recomendável criar um gerador `rng` e usar os métodos dele para gerar números pseudoaleatórios: `rng.metodo1()`, `rng.metodo2()` [em vez de simplesmente usar `np.random.metodo1()`, `np.random.metodo2()`] pois fica mais fácil de reproduzir os resultados.

### Distribuição uniforme (contínua)

Gera valores onde qualquer ponto dentro do intervalo $[a, b)$ tem a mesma probabilidade de ser sorteado.

```python
# 5 valores aleatórios entre 10.0 e 20.0
amostra_uniforme = rng.uniform(low=10.0, high=20.0, size=5)
print(amostra_uniforme)
```

```python
print(rng.uniform(low=10.0, high=20.0, size=5)) # cada vez que executamos gera valores diferentes!
```

A sequência é sempre a mesma a partir do momento em que o objeto `rng` foi criado.

Por exemplo, se criarmos outro gerador com a mesma semente e fizermos a mesma amostragem, obteremos exatamente a array em `amostra_uniforme`.

```python
outro_rng = np.random.default_rng(seed=42)
primeira_amostra_uniforme = outro_rng.uniform(low=10.0, high=20.0, size=5)
print(primeira_amostra_uniforme)
```

Se amostrarmos de novo, obteremos algo previsível... (o mesmo de quando executamos rng.uniform pela segunda vez).

```python
print(outro_rng.uniform(low=10.0, high=20.0, size=5))
```

Sem usar gerador explícito:

```python
np.random.seed(42) # gerador implícito inicializado com seed 42
np.random.uniform(low=10.0, high=20.0, size=5) # 1º chamada ao gerador implícito: então fará sempre a mesma escolha
```

Temos menos controle, pois a escolha pseudoaleatória depende de TODAS as chamadas anteriores ao gerador implícito do `np.random` desde que esse gerador implícito foi inicializado. Geralmente a inicialização `np.random.seed(...)` acontece só no início do código, e sabe-se lá o número de chamadas que foram feitas. Gerador explícito permite que seu experimento seja reprodutível mesmo que ele ocorra em paralelo (basta cada 'thread' usar um gerador próprio).

### Distribuição normal (gaussiana)

Gera valores conforme uma distribuição normal de média $\mu$ e desvio padrão $\sigma$.

```python
# Simulando a altura (em cm) de uma população: média 170 cm, desvio padrão 8 cm
media_teorica = 170.0
desvio_teorico = 8.0
tamanho_amostra = 1000
```

```python
alturas = rng.normal(media_teorica, desvio_teorico, size=100)
print(alturas)
print('shape:', alturas.shape)
```

Inspecionando os primeiros 5 indivíduos gerados:

```python
print(alturas[:5])
```

Pela Lei dos Grandes Números, a média e o desvio da amostra simulada devem convergir para os valores teóricos configurados.

```python
print("Média empírica observada:", alturas.mean())
```

```python
print("Desvio padrão empírico observado:", alturas.std())
```

### Distribuição binomial

Modela a contagem de sucessos em $n$ experimentos independentes, cada um com probabilidade $p$.

Cenário: 10 pacientes recebem um medicamento com eficácia de 70% ($p = 0.7$). Simulamos esse ensaio 5 vezes.

```python
sucessos = rng.binomial(n=10, p=0.7, size=5)
```

```python
print("Número de sucessos em cada ensaio:", sucessos)
```

### Amostragem aleatória com choice

O método `choice` sorteia elementos de um vetor existente (equivalente a uma amostragem estatística).

```python
grupos = np.array(["Controle", "Tratamento A", "Tratamento B"])
```

Sorteio com reposição (o mesmo elemento pode ser selecionado várias vezes):

```python
sorteio_com_reposicao = rng.choice(grupos, size=6, replace=True)
print(sorteio_com_reposicao)
```

```python
sorteio_sem_reposicao = rng.choice(grupos, size=2, replace=False)
print(sorteio_sem_reposicao)
```

### Distribuição categórica

Podemos associar a probabilidade teórica de ocorrência a cada categoria usando o argumento `p` (a soma dos pesos deve ser 1.0).

```python
tipos_sanguineos = np.array(["O", "A", "B", "AB"])
pesos = [0.45, 0.40, 0.12, 0.03]
```

```python
amostra_sangue = rng.choice(tipos_sanguineos, size=10, p=pesos)
print(amostra_sangue)
```

Obs.: se você tem uma lista arbitrária de pesos e quer fazer a soma dar 1.0, basta dividir pela soma dos pesos.

```python
pesos_arbitrarios = np.array([1, 1, 1, 2, 3])
pesos_somando_1 = pesos_arbitrarios / pesos_arbitrarios.sum()
print(pesos_somando_1)
```

### Embaralhamento e permutações

O método `permutation` retorna uma nova sequência reordenada aleatoriamente sem alterar o array original.

```python
indices_embaralhados = rng.permutation(indices)
print(indices_embaralhados)
```

```python
print("Original:", indices)
print("Permutado:", indices_embaralhados)
```
