# Aula 20: estruturas de dados - dicionários

Nesta aula, vamos estudar os dicionários em Python. Ao contrário das listas e tuplas, que são indexadas por posições numéricas de 0 a N - 1 (em que N é o tamanho da sequência), os dicionários são indexados por **chaves** quaisquer imutáveis.

## 1. Conceito e crição de dicionários

Um dicionário mapeia chaves únicas para seus respectivos valores.
As chaves precisam ser tipos imutáveis (como strings, inteiros ou tuplas).
Os valores podem ser de qualquer tipo de dado.

```python
# criando dicionários vazios
dicionario_vazio_1 = {}
dicionario_vazio_2 = dict()

print(type(dicionario_vazio_1))
```

```python
# criação com chaves do tipo string e valores variados
aluno = {
    "nome": "Pâmela",
    "idade": 20,
    "curso": "Estatística",
    "matriculado": True
}

print(aluno)
```

```python
# criação usando dict()
precos = dict(arroz=5.50, feijao=8.20, macarrao=4.30)
print(precos)
```

## 2. Acesso aos elementos

Podemos acessar um valor informando a chave entre colchetes.
Se a chave não existir, o Python gera um erro de chave (KeyError).

```python
# acesso direto por chave
print(aluno["nome"])
print(aluno["curso"])
```

Para evitar erros ao buscar chaves que podem nao existir, usamos o metodo get().
O metodo get() retorna None ou um valor padrao se a chave nao for encontrada.

```python
aluno["nota"]
```

```python
# acesso seguro usando get
nota = aluno.get("nota")
print(nota)

# get com valor padrao definido pelo usuario
nota_com_padrao = aluno.get("nota", 0.0)
print(nota_com_padrao)
```

## 3. Inserção e modificação de dados

Dicionários sao **mutáveis** (assim como listas).
Para alterar um valor existente ou criar uma nova chave, usamos a atribuição com colchetes.

```python
aluno = { "nome": "Pâmela", "idade": 20, "curso": "Estatística", "matriculado": True }
```

```python
# alterando um valor existente
aluno["idade"] = 21
print(aluno)
```

```python
# adicionando uma nova chave
aluno["semestre"] = 3
print(aluno)
```

```python
# atualização em lote com o método update()
novidades = {"cidade": "Fortaleza", "matriculado": False}
aluno.update(novidades)
print(aluno)
```

## 4. Remoção de elementos

Podemos remover itens de um dicionario usando
- comando `del`
- metodo `pop()`
- metodo `clear()`

```python
aluno = {'nome': 'Pâmela', 'idade': 21, 'curso': 'Estatística', 'matriculado': False, 'semestre': 3, 'cidade': 'Fortaleza'}
```

```python
# remocao usando o comando del
del aluno["matriculado"]
print(aluno)
```

```python
# remocao usando o metodo pop(), que tambem retorna o valor removido
semestre_removido = aluno.pop("semestre")
print("removido:", semestre_removido)
print(aluno)
```

```python
# esvaziando o dicionario completamente
aluno.clear()
print(aluno)
```

## 5. Verificando a existencia de chaves

O operador "in" permite verificar se uma determinada chave esta presente no dicionario.

```python
aluno = {'nome': 'Pâmela', 'idade': 21, 'curso': 'Estatística', 'matriculado': False, 'semestre': 3, 'cidade': 'Fortaleza'}
```

```python
print("nome" in aluno)
print("nota" in aluno)
```

## 6. Iteração em dicionarios

A seguir, 3 formas de percorrer dicionários:
- apenas pelas chaves
- apenas pelos valores
- por ambos simultaneamente (pares chave-valor)

```python
aluno = {'nome': 'Pâmela', 'idade': 21, 'curso': 'Estatística', 'matriculado': False, 'semestre': 3, 'cidade': 'Fortaleza'}
```

```python
# iteração padrão (percorre as chaves)
for chave in aluno:
    print(chave)
```

```python
# iteração explicita usando o metodo keys()
for chave in aluno.keys():
    valor = aluno[chave]
    print(f"{chave} -> {valor}")
```

```python
# iteração usando o método values()
for valor in aluno.values():
    print(valor)
```

```python
# iteração usando o método items()
for chave, valor in aluno.items():
    print(f"{chave}: {valor}")
```

## 7. Exemplo: contagem de frequência

Dicionários são ideais para contar a frequência de ocorrência de elementos em uma coleção.

```python
palavras = ["sol", "mar", "ceu", "sol", "lua", "mar", "sol"]
contagem = {}

for palavra in palavras:
    if palavra in contagem:
        contagem[palavra] += 1
    else:
        contagem[palavra] = 1

for termo, qtd in contagem.items():
    print(f"{termo}: {qtd}")
```

## 8. Estruturas de dados: conjuntos (sets)

Conjuntos são coleções **desordenadas** de elementos únicos (não aceitam valores duplicados).
Assim como as chaves dos dicionários, os conjuntos são baseados em tabelas hash, o que torna a verificação de pertencimento (operador `in`) extremamente rápida.

```python
# criação de conjuntos com valores
numeros = {1, 2, 3, 4, 4, 2, 1}
print(numeros)
```

```python
# atenção: {} cria um dicionário vazio. Para conjunto vazio, usamos set()
conjunto_vazio = set()
print(type(conjunto_vazio))
```

## 9. Conversão e remoção de duplicatas

O uso mais comum de conjuntos no dia a dia da programação é eliminar elementos duplicados de listas.

```python
lista_com_duplicados = [10, 20, 10, 30, 20, 40, 50, 10]

conjunto_intermediario = set(lista_com_duplicados) # ORDEM PERDIDA NESTE MOMENTO
print(conjunto_intermediario)

lista_sem_duplicados = list(conjunto_intermediario)
print(lista_sem_duplicados)
```

Conjuntos não dão garantia de ordem. Por isso, os elementos podem aparecer em qualquer posição ao converter para lista!

Se a ordem for importante, use `dict.fromkeys` em vez de `set`. Isso funciona porque desde Python 3.7, dicionários preservam a ordem de inserção das chaves.

```python
lista_com_duplicados = [10, 20, 10, 30, 20, 40, 50, 10]

dicionario_intermediario = dict.fromkeys(lista_com_duplicados)
print(dicionario_intermediario)

lista_sem_duplicados = list(dicionario_intermediario)
print(lista_sem_duplicados)
```

## 10. Operações básicas de conjuntos

Conjuntos são **mutáveis** e permitem adição e remoção de itens de forma direta.

```python
cores = {"azul", "verde", "vermelho"}

# adicionar elemento
cores.add("amarelo")

# remover elemento
cores.remove("verde")

print(cores)
print("azul" in cores) # Operação BEM mais rápida do que em listas quando a coleção é grande!
```

## 11. Operações matemáticas de conjuntos

O Python permite realizar operações clássicas da teoria dos conjuntos de forma bastante expressiva:
- união (|): elementos que estão em pelo menos um dos conjuntos
- interseção (&): elementos que estão em ambos os conjuntos
- diferença (-): elementos do primeiro conjunto que não estão no segundo

```python
curso_a = {"Ana", "Bruno", "Carlos", "Diana"}
curso_b = {"Carlos", "Diana", "Eduardo", "Fernanda"}

# união
print("união:", curso_a | curso_b)

# interseção
print("interseção:", curso_a & curso_b)

# diferença (alunos apenas no curso a)
print("apenas curso a:", curso_a - curso_b)
```

## 12. Iteração em conjuntos

Você pode percorrer um conjunto, mas NÃO HÁ GARANTIA DE ORDEM.

```python
curso_a = {"Ana", "Bruno", "Carlos", "Diana"}
for x in curso_a:
    print(x)
```

```python
list({"Ana", "Bruno", "Carlos", "Diana"}) # Conjunto (sem garantia de ordem) --> Lista
```

### Exercício 1

Dado o dicionário abaixo, mostre apenas os nomes dos alunos (linha a linha) que tiraram nota maior ou igual a 7.0.

```python
notas = {"Ana": 8.5, "Bruno": 6.0, "Carla": 9.2, "Diego": 4.5}

for aluno, nota in notas.items():
    if nota < 7.0:
        continue
    print(aluno)
```

### Exercício 2

Escreva um programa que usa um dicionário para contar quantas vezes cada caractere aparece no texto.

```python
texto = "conte a frequência de caracteres nesta frase (inclusive acentos, símbolos, etc.)"

freq = {}

for c in texto:
    if c not in freq:
        freq[c] = 1
    else:
        freq[c] += 1

for k,v in freq.items():
    print(f'Caractere ({k}): {v}')
```
