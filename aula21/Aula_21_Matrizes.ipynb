# Aula 21: matrizes e estruturas bidimensionais

Nesta aula, vamos estudar como representar e manipular dados organizados em tabelas (linhas e colunas).

## 1. Representação de matrizes

Uma matriz é uma estrutura bidimensional composta por linhas e colunas.
Em Python, cada linha da matriz é representada como uma lista interna contida em uma lista externa.

```python
# matriz 3x3 (3 linhas e 3 colunas)
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(matriz)
```

## 2. Indexação dupla

Para acessar um elemento específico, informamos dois índices entre colchetes:
- o primeiro colchete seleciona a linha: matriz[linha]
- o segundo colchete seleciona a coluna: matriz[linha][coluna]

Lembre-se de que a contagem de índices em Python sempre começa em zero.

```python
# acessando uma linha inteira
print("primeira linha:", matriz[0])
```

```python
# acessando elementos pontuais
print("elemento na linha 0, coluna 0:", matriz[0][0])
print("elemento na linha 1, coluna 2:", matriz[1][2])
```

Podemos modificar diretamente o valor de qualquer posição da matriz utilizando a atribuição direta com o par de índices.

```python
# alterando o elemento central da matriz
matriz[1][1] = 99
print("linha 1 modificada:", matriz[1])
```

## 3. Varredura com laços aninhados

Para percorrer todos os elementos de uma matriz, usamos dois laços for aninhados:
- o laço externo percorre as linhas
- o laço interno percorre as colunas de cada linha

```python
# percorrendo diretamente pelos elementos
for aluno in matriz:
    for elemento in aluno:
        print(elemento)
```

Quando precisamos saber as coordenadas exatas (linha e coluna) de cada elemento, usamos a função range() combinada com len().

```python
total_linhas = len(matriz)
total_colunas = len(matriz[0])

for aluno in range(total_linhas):
    for elm in range(total_colunas):
        print(f"posição [{aluno}][{elm}] = {matriz[aluno][elm]}")
```

## 5. Criação dinâmica de matrizes

Para construir uma matriz arbitrária, criamos cada linha individualmente e a adicionamos à lista principal com o método append().

```python
linhas = 3
colunas = 4
matriz = []

# CONSTRUIR MATRIZ 3 x 4 EM QUE A[i][j] = i * 4 + j
for aluno in range(linhas):
    nova_linha = []
    for elm in range(colunas):
        nova_linha.append(aluno * 4 + elm)
    matriz.append(nova_linha)

print(matriz)
```

## 6. Mutabilidade

Atenção! Listas são mutáveis, então cuidado ao adicionar listas a uma matriz:

```python
linha_1 = [0, 1]
linha_2 = linha_1 # AQUI NÃO FOI FEITA UMA CÓPIA... É A MESMA LISTA NA MEMÓRIA!
matriz_estranha = [linha_1, linha_2]
print(matriz_estranha)
```

```python
matriz_estranha[0][0] = 99
print(matriz_estranha)
```

Isso acontece se você usar repetição `*` para inicializar uma matriz (NÃO RECOMENDADO):

```python
M = 4
N = 3
matriz_zeros_m_por_n = [[0] * N] * M
print(matriz_zeros_m_por_n)
```

```python
matriz_zeros_m_por_n[0][0] = 1
```

```python
print(matriz_zeros_m_por_n)
```

Isso também acontece se você adicionar a mesma linha a duas matrizes distintas:

```python
linha_compartilhada = [15, 30]
matriz_A = [[0, 1], [1, 2]]
matriz_B = [[3, 4], [5, 6]]

matriz_A.append(linha_compartilhada)
matriz_B.append(linha_compartilhada)
print(matriz_A)
print(matriz_B)
```

```python
matriz_A[2][0] = 9999
print(matriz_A)
print(matriz_B)
```

Para resolver esses problemas, faça uma cópia:

```python
lista_original = [1000, 2000]
lista_copia = lista_original.copy() # usando .copy()
```

```python
lista_original = [1000, 2000]
lista_copia = lista_original[:] # ou usando fatiamento [:]
```

```python
# agora você pode passar a original para uma matriz A e a cópia para a matriz B!linha_compartilhada = [15, 30]
matriz_A = [[0, 1], [1, 2]]
matriz_B = [[3, 4], [5, 6]]

matriz_A.append(lista_original)
matriz_B.append(lista_copia)
print(matriz_A)
print(matriz_B)
```

```python
matriz_A[2][0] = 9999
print(matriz_A)
print(matriz_B)
```

### Exercício 1

Dada a matriz numérica a seguir, utilize laços aninhados para percorrer todos os seus valores e calcular a soma total de todos os elementos contidos nela.

```python
matriz_numeros = [
    [10, 15, 20],
    [5, 25, 30],
    [8, 12, 14]
]

soma = 0
for aluno in matriz_numeros:
    for elm in aluno:
        soma += elm
print(soma)
```

### Exercício 2

Dada a matriz quadrada a seguir (mesmo número de linhas e colunas), exiba apenas os valores que pertencem à diagonal principal (onde o índice da linha é igual ao índice da coluna) e calcule a soma desses valores (chamado [traço](https://pt.wikipedia.org/wiki/Tra%C3%A7o_(%C3%A1lgebra_linear)) da matriz).

{9FDFFB30-00E5-4FAF-B000-C89912A94C87}.png

```python
matriz_quadrada = [
    [4, 8, 1, 9],
    [3, 7, 2, 5],
    [6, 1, 8, 3],
    [2, 4, 9, 5]
]

soma = 0
for i, aluno in enumerate(matriz_quadrada):
    for j, elm in enumerate(aluno):
        if i != j:
            continue
        soma += elm
print(soma)
```

### Exercício 3

A matriz a seguir armazena as notas de três provas de quatro alunos diferentes (cada linha representa um aluno). Percorra cada linha, calcule e exiba a média aritmética de cada aluno.

```python
notas_alunos = [
    [7.5, 8.0, 9.0],
    [5.0, 6.0, 4.5],
    [9.0, 9.5, 10.0],
    [6.0, 7.0, 6.5]
]

for i, aluno in enumerate(notas_alunos):
    soma = 0
    for j, nota in enumerate(aluno):
        soma += nota
    print(f'Média do aluno {i}: {soma/3:.2f}')

```
