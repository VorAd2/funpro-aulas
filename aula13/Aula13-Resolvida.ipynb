# Aula 13: Controle de Laços: sentinelas, `break` e `continue`

Nas aulas anteriores, aprendemos a repetir blocos de código com repetição condicional (`while`) e repetição contada (`for`).

Muitas vezes precisamos de controle mais refinado sobre o fluxo:
* **Interromper** um laço imediatamente ao encontrar uma resposta ou valor de parada.
* **Pular** o restante da iteração atual caso um dado seja inválido ou irrelevante.
* **Aguardar** um sinal do usuário informando que os dados terminaram (valor sentinela).

Nesta aula, aprenderemos a utilizar **sentinelas**, o comando **`break`**, o comando **`continue`** e o comportamento especial da cláusula **`else` em laços**.

## 1. Valores sentinela

Um **valor sentinela** (também chamado de *flag de parada* ou *sinalizador*) é um valor pré-definido, fora do conjunto normal de dados válidos, cuja única finalidade é sinalizar ao programa que a entrada de dados chegou ao fim.

Imagine um caixa de supermercado passando produtos. O caixa não sabe se o cliente comprou 3 ou 40 itens. Ele apenas continua bipando até apertar o botão "Finalizar compra". Esse comando de finalizar é a **sentinela**.

### Sentinela com `while`
1. Faz-se uma leitura antes do laço (leitura de inicialização).
2. O laço testa se o valor lido é diferente da sentinela.
3. Ao final do bloco do laço, faz-se uma nova leitura.

```python
print("Somador de números")

soma = 0.0
sentinela = 0
numero = float(input(f"Digite um número (ou {sentinela} para sair): "))

while numero != sentinela:
    soma += numero
    numero = float(input(f"Digite o próximo número (ou {sentinela} para sair): "))

print(f"\nSoma total dos valores digitados: {soma:.2f}")
```

## 2. O comando `break`

No código anterior, fomos obrigados a duplicar a linha de `input`: uma antes do laço e outra dentro dele. Essa repetição de código pode gerar erros caso precisemos alterar a mensagem ou a lógica de leitura.

Para resolver isso, o Python oferece o comando **`break`** (que significa "interromper" ou "quebrar").

### Como o `break` funciona:
* Ao ser executado dentro de um `while` ou `for`, o `break` encerra o laço **imediatamente**.
* A execução do programa salta direto para a primeira linha após o bloco do laço.
* Quaisquer instruções que estiverem abaixo do `break` dentro do laço são ignoradas.

### `while True`
Ao combinar `while True:` com uma condição que aciona o `break`, lemos o dado uma única vez dentro do laço e decidimos a saída no momento em que ele é lido:

```python
soma = 0.0
sentinela = 0

print(f"Somador de números")

while True:
    numero = float(input(f"Digite um número (ou {sentinela} para sair): "))
    if numero == sentinela:
        print("Sentinela detectada. Encerrando o laço...")
        break  # Sai imediatamente do while
    soma += numero

print(f"Soma final calculada: {soma:.2f}")
```

### 2.1. O comando `break` em laços `for`

O `break` não serve apenas para `while`. Ele é amplamente utilizado em laços `for`, principalmente em **operações de busca**.

Se estivermos procurando um elemento específico em um intervalo e já o encontramos, continuar iterando seria um desperdício desnecessário de processamento.

```python
# Busca antecipada: encontrar o primeiro número divisível por 7 e 11
# em um intervalo entre 1 e 200

alvo = None

for n in range(1, 201):
    if n % 7 == 0 and n % 11 == 0:
        alvo = n
        print(f"Número encontrado: n = {n}")
        break  # Interrompe o laço; não precisamos testar até 200!

print(f"Busca encerrada. alvo = {alvo}")
```

## 3. O comando `continue`

Enquanto o `break` abandona o laço por completo, o comando **`continue`**

* interrompe apenas a iteração atual,
* faz o laço saltar imediatamente para a iteração seguinte (reavaliando a condição no `while` ou pegando o próximo elemento no `for`).

| Comando | O que faz com o laço? |
| :--- | :--- |
| `break` | Finaliza o laço por completo. |
| `continue` | Finaliza a iteração atual e avança para a próxima. |

> **Atenção com o `while`:** se você usar `continue` em um laço `while`, garanta que a atualização da variável de controle (ex: `contador += 1`) ocorra **antes** do `continue`. Caso contrário, o programa cairá em um **loop infinito**, pois a variável nunca mudará de valor!

```python
print("Números de 1 a 10 (exceto múltiplos de 3):")
soma = 0
for i in range(1, 11):
    if i % 3 == 0:
        print(f"-> Pulando o {i}...")
        continue  # Vai direto para o próximo i
    soma += i
    print(f"Número: {i}")
print("Soma final:", soma)
```

```python
contador = 0
while contador < 10:
    contador += 1  # Incremento ANTES do teste de continue
    print(f"contador é {contador}")

    if contador % 2 == 0:
        continue  # Pula os pares

    print(f"Ímpar encontrado: {contador}")
```

### 3.1. Usando `continue` para filtrar entradas inválidas

O `continue` é excelente para implementar a estratégia de "guarda": quando uma entrada for inválida, nós avisamos o usuário e chamamos `continue` para reiniciar o ciclo, evitando blocos `if-else` aninhados.

```python
# Acumular 5 números estritamente positivos fornecidos pelo usuário
total_positivos = 0
soma = 0.0

while total_positivos < 5:
    valor = float(input(f"Digite o {total_positivos + 1}º número positivo: "))

    if valor <= 0:
        print("Valor inválido! O número deve ser maior que zero. Tente novamente.")
        continue  # Volta ao início do while sem somar nem contar

    soma += valor
    total_positivos += 1

print(f"\nSucesso! 5 números computados. Soma total: {soma:.2f}")
```

## 4. A cláusula `else` em Laços (Recurso exclusivo do Python)

Em Python, tanto o `while` quanto o `for` podem ter uma cláusula `else` opcional ao final.

A regra fundamental do `else` em laços é:
* O bloco do `else` **só é executado se o laço terminar naturalmente** (ou seja, quando a condição do `while` se tornar falsa ou quando o `for` percorrer toda a sequência).
* Se o laço for interrompido por um **`break`**, o bloco `else` é **ignorado**.

Isso é muito útil em buscas, eliminando a necessidade de criar variáveis auxiliares (flags) para checar se o elemento foi ou não encontrado.

```python
# Exemplo: Testando se um número N é primo usando for ... break ... else
# Um número primo só é divisível por 1 e por ele mesmo.
N = int(input("Digite um número inteiro maior que 1: "))

if N <= 1:
    print("Por definição, números <= 1 não são primos.")
else:
    quantidade_testada = 0
    for divisor in range(2, N):
        quantidade_testada += 1
        if N % divisor == 0:
            print(f"{N} NÃO é primo (é divisível por {divisor}).")
            break  # Interrompe o laço; o 'else' NÃO será executado!
    else:
        # Só executa se o for percorrer de 2 até N-1 SEM acionar o break
        print(f"Parabéns! {N} É um número primo.")
    print(f"{quantidade_testada} números foram testados para descobrir isso") # executa com ou sem break acionado
```

## 5. Juntando tudo

Vamos construir um sistema interativo de caixa de loja que aplica todos os conceitos da aula:
1. Utiliza `while True` para ler itens indefinidamente.
2. Usa a sentinela `0` com `break` para encerrar a compra.
3. Usa `continue` para descartar e alertar sobre valores negativos inseridos por engano.
4. Acumula o valor total da compra e a quantidade de produtos registrados.

```python
sentinela = 0

print("========================================")
print("     SISTEMA DE CAIXA REGISTRADORA      ")
print("  Digite o preço de cada produto.       ")
print(f"  Digite {sentinela} no valor para fechar a conta.")
print("========================================")

total_conta = 0.0
qtd_itens = 0

while True:
    preco = float(input(f"Preço do item #{qtd_itens + 1} (R$): "))

    # 1. Checagem da sentinela de encerramento
    if preco == sentinela:
        if qtd_itens == 0:
            print("Nenhum item foi registrado. Encerrando operação.")
            break
        print("\nFechando conta...")
        break

    # 2. Filtragem de entrada inválida com continue
    if preco < 0:
        print(">> Erro: Preço não pode ser negativo! Item descartado.")
        continue

    # 3. Processamento do item válido
    total_conta += preco
    qtd_itens += 1
    print(f"Item #{qtd_itens} adicionado! Subtotal: R$ {total_conta:.2f}")

if qtd_itens > 0:
    print()
    print("============== RESUMO ==============")
    print(f"Total de itens comprados: {qtd_itens}")
    print(f"Valor final a pagar:     R$ {total_conta:.2f}")
    print(f"Valor médio por item:    R$ {(total_conta / qtd_itens):.2f}")
    print("====================================")
```

### Exercício 1: Teste de mesa

Sem executar o código abaixo no computador, trace mentalmente ou em um papel os valores das variáveis `x` e `total` em cada repetição.

Após responder, execute a célula para validar seu raciocínio.

```python
# Exercício 1: Teste de mesa
total = 0
x = 0

while x < 15:
    x += 1

    if x % 4 == 0:
        continue

    if x > 10:
        break

    total += x
    print(f"Somado: {x}")

print(f"Total final: {total}")
```

### Exercício 2: Somador com sentinela "fim"

Crie um programa que peça números ao usuário continuamente.
* O usuário pode digitar números reais (`float`) ou digitar a palavra `"fim"` (sentinela) para parar.
* Se o usuário digitar `"fim"`, o programa deve interromper o laço (`break`) e exibir:
  1. A quantidade de números válidos digitados.
  2. A soma de todos os números.
  3. A média aritmética dos números (evite erro de divisão por zero caso o usuário digite `"fim"` logo na primeira vez!).


```python
# Sua solução aqui

# Vamos assumir um usuário ideal que fornecerá exclusivamente um número ou 'fim'.
qtt = 0
summ = 0
while True:
    data = input("Type a real number (or 'fim' to break): ")
    if data == 'fim':
        break
    
    number = float(data)
    qtt += 1
    summ += number

print(f"Quantity: {qtt}")
print(f"Sum: {summ}")
print('No numbers' if qtt == 0 else f'Arith: {summ / qtt}')
```

### Exercício 3: Média de Notas com Filtro e Sentinela

Escreva um programa que solicite notas escolares aos alunos (de `0.0` a `10.0`):
* O programa deve parar imediatamente quando o usuário digitar a sentinela `-1`.
* Caso o usuário digite um valor que não seja `-1` e esteja fora do intervalo permitido (por exemplo, `15.0` ou `-4.0`), exiba a mensagem `"Nota inválida! Ignorada."` e utilize o comando `continue` para não registrar essa nota.
* Ao final da execução (após o `-1`), informe a quantidade de notas válidas computadas e a média da turma.

```python
# Sua solução aqui

# Vamos assumir um usuário ideal que digitará apenas números.

qtt_validate = 0
summ = 0

while True:
    number = float(input('Type the number: '))
    if number == -1:
        break

    if (number < 0.0 or number > 10.0):
        print('Nota inválida! Ignorada')
        continue

    qtt_validate += 1
    summ += number

print(f'Qtt validate: {qtt_validate}')
print('No numbers' if qtt_validate == 0 else f'Average: {summ / qtt_validate}')
```

### Exercício 4: Sistema de Login com Limite de Tentativas

Simule a tela de autenticação de um sistema bancário:
* O usuário tem até **3 tentativas** para acertar a senha correta (a senha cadastrada é `"python123"`).
* Utilize um laço `for` com `range(1, 4)`.
* Se o usuário acertar a senha: exiba `"Acesso concedido!"` e saia do laço imediatamente com `break`.
* Se errar, avise quantas tentativas ainda restam.
* Utilize a cláusula `else` do laço para exibir `"Conta bloqueada por excesso de tentativas!"` caso o laço chegue ao fim sem nenhum `break`.

```python
# Sua solução aqui

password = 'python123'
lucky = 3

for _ in range(1,4):
    attempt = input('Type your password: ')
    if attempt != password:
        lucky -= 1
        print('Wrong password' if lucky == 0 else f'Wrong password. Try again ({lucky} attempts remaining)\n')
        continue
    print('Acesso concedido')
    break
else:
    print('Conta bloqueada por excesso de tentativas!')
```

### Exercício 5: Encontrando o Menor Número Positivo

Crie um programa que leia inteiros digitados pelo usuário até que o número `0` seja fornecido (sentinela com `break`).

O programa deve:
* Ignorar qualquer número negativo digitado usando `continue`.
* Ao final da leitura, informar qual foi o **menor número estritamente positivo** digitado.
* Se nenhum número positivo tiver sido digitado, informe isso ao usuário.

*Dica:* Inicialize uma variável `menor = float('inf')`, que representa um número maior que qualquer número finito.

```python
# Sua solução aqui

# Vamos assumir um usuário ideal que digitará apenas números inteiros.

min = float('inf')
validate_user = False

while True:
    integer = int(input('Type the integer: '))
    if integer == 0:
        break
    if integer < 0:
        continue

    print('Válido digitado:', integer)
    validate_user = True
    if integer < min:
        min = integer

print('Nenhum número válido digitado' if validate_user == False else f'Menor: {min}')

```
