# Aula 18: Tuplas e manipulação de strings

Nas aulas anteriores, exploramos listas (sequências mutáveis). Hoje estudaremos estruturas de dados similares, mas imutáveis:

- **Tuplas:** coleções sequenciais imutáveis de dados quaisquer.
- **Strings:** sequências imutáveis de caracteres, com métodos próprios para manipulação de texto.

## 1. O que é uma tupla?

Uma tupla é uma coleção sequencial de elementos protegida contra modificações. Assim como as listas, ela pode guardar qualquer tipo de dado (números, textos, booleanos ou outras coleções).

Principais características:
- **Imutável:** após criada, você não pode adicionar, remover ou trocar elementos.
- **Eficiência e segurança:** ocupam menos memória que listas e garantem que dados não sejam alterados acidentalmente durante o programa.

[Sintaxe] delimitada por parênteses `()`, separada por vírgulas.
- Exemplo: `(2, 4, 3)`.


### Tupla vazia

```python
tupla_vazia_1 = ()
tupla_vazia_2 = tuple()

print("Tupla vazia:", tupla_vazia_1)
print("Tipo:", type(tupla_vazia_1))
```

### Tupla com múltiplos elementos

```python
dias_semana = ("segunda", "terça", "quarta", "quinta", "sexta", "sábado", "domingo")
ponto_gps = (-23.5505, -46.6333)
usuario = ("Mariana", 24, True)

print("Dias da semana:", dias_semana)
print("Coordenadas:", ponto_gps)
print("Dados do usuário:", usuario)
```

### Atenção: tupla de 1 elemento

Atenção ao criar uma tupla de um único elemento: a vírgula é obrigatória. Sem a vírgula, o Python interpreta os parênteses apenas como o indicador de precedência (no caso de 1 elemento, sempre redundante).

```python
# Sem vírgula: o Python considera apenas um número inteiro
nao_eh_tupla = (42)
print("Tipo sem vírgula:", type(nao_eh_tupla))
```

```python
# Com vírgula: agora é uma tupla
eh_uma_tupla = (42,)
print("Tipo com vírgula:", type(eh_uma_tupla))
```

Também é possível criar tuplas omitindo os parênteses, usando apenas as vírgulas. Essa prática é chamada de empacotamento de tupla (*tuple packing*).

```python
dados = 10, 20, 30
print("Dados empacotados:", dados)
print("Tipo:", type(dados))
```

## 2. Indexação, fatiamento e a imutabilidade

O acesso a posições individuais e os recortes com fatiamento funcionam exatamente da mesma forma que nas listas.

```python
meses = ("janeiro", "fevereiro", "março", "abril", "maio")

# Indexação: meses[index]
print("Primeiro mês:", meses[0])
print("Último mês:", meses[-1])
print("Penúltimo mês:", meses[-2])
```

```python
# Fatiamento: meses[inicio:fim:passo]
print("Primeiros três meses:", meses[:3])
print("Do índice 2 até o final:", meses[2:])
print("Invertendo a tupla:", meses[::-1])
```

A grande diferença em relação às listas é que tuplas **NÃO** aceitam alterar elementos.

```python
cores = ("vermelho", "verde", "azul")
cores[0] = "amarelo" # erro!
```

Note que as operações de indexação (`dado = meses[3]`) e fatiamento (`nova = meses[::-1]`) NÃO alteram o objeto original (apenas geram um novo) e portanto são aplicáveis a tuplas.

## 3. Desempacotamento de tuplas (*tuple unpacking*)

Podemos usar indexação para acessar cada elemento de uma tupla. Faríamos:

```python
ponto = (15, 30)
x = ponto[0]
y = ponto[1]
print("Coordenada x:", x)
print("Coordenada y:", y)
```

Alternativamente, podemos usar **desempacotamento**, que é um atalho sintático para extrair os elementos de uma tupla para variáveis.

Basta fazer uma atribuição em que o ALVO (lado esquerdo) são variáveis separadas por vírgula. O número de variáveis deve ser idêntico ao número de elementos contidos na tupla.

```python
ponto = (15, 30)
x, y = ponto
print("Coordenada x:", x)
print("Coordenada y:", y)
```

```python
ponto = (15, 30)
(x, y) = ponto # também pode usar parênteses, se quiser
print("Coordenada x:", x)
print("Coordenada y:", y)
```

```python
# Exemplo: troca de valores entre duas variáveis sem variável auxiliar
a = 5
b = 10
print(f"Antes da troca: a = {a}, b = {b}")

a, b = b, a
print(f"Depois da troca: a = {a}, b = {b}")
```

Isso também funciona com listas...

```python
ponto = [15, 30]
x, y = ponto
print("Coordenada x:", x)
print("Coordenada y:", y)
```

### Desempacotamento com asterisco *

Se você quiser capturar apenas alguns elementos e agrupar o restante em uma lista, pode usar o operador asterisco `*`.

```python
notas_aluno = (10, 8.5, 7.0, 9.5, 6.0)

primeira, *intermediarias, ultima = notas_aluno

print("Primeira nota:", primeira)
print("Notas intermediárias (viram lista):", intermediarias)
print("Última nota:", ultima)
```

Isso acima seria QUASE equivalente ao código abaixo

```python
notas_aluno = (10, 8.5, 7.0, 9.5, 6.0)

primeira = notas_aluno[0]
intermediarias = notas_aluno[1:-1]
ultima = notas_aluno[-1]

print("Primeira nota:", primeira)
print("Notas intermediárias (aqui continua tupla):", intermediarias)
print("Última nota:", ultima)
```

A diferença é que, no último código, `intermediarias` continuou como tupla pelo fatiamento.

Pode-se converter tupla em lista e lista em tupla usando os métodos `list` e `tuple` respectivamente:

```python
lista_exemplo1 = [1, 2, 3]
tupla_exemplo1 = tuple(lista_exemplo1)

tupla_exemplo2 = (3, 2, 1)
lista_exemplo2 = list(tupla_exemplo2)

print(lista_exemplo1, tupla_exemplo1)
print(tupla_exemplo2, lista_exemplo2)
```

## 4. Métodos e funções builtin com tuplas

Como não podemos alterar tuplas, métodos como `append`, `insert`, `pop`, `remove` e `sort` não existem nelas. Tuplas possuem apenas métodos de consulta:
- `count(item)`: informa quantas vezes um item aparece.
- `index(item)`: descobre em qual índice o item está pela primeira vez.

As funções builtin (nativas) `len()`, `min()`, `max()`, `sum()` e o operador `in` continuam funcionando normalmente.

```python
valores = (5, 2, 8, 2, 9, 2, 4)

print("Tamanho total (len):", len(valores))
print("Soma total (sum):", sum(valores))
print("Menor valor (min):", min(valores))
print("Maior valor (max):", max(valores))

print("Quantas vezes o número 2 aparece?:", valores.count(2))
print("Em qual posição o número 8 aparece?:", valores.index(8))
print("O número 9 está na tupla?:", 9 in valores)
```

Caso seja necessário alterar dados de uma tupla, pode-se convertê-la em lista, fazer as modificações necessárias e depois convertê-la de volta em tupla.

```python
tupla_original = ("A", "B", "C")
print("Original:", tupla_original)

# Converte para lista
lista_temporaria = list(tupla_original)
lista_temporaria.append("D") # agora é ["A", "B", "C", "D"]

# Converte de volta para tupla
tupla_atualizada = tuple(lista_temporaria) # agora é ("A", "B", "C", "D")
print("Atualizada:", tupla_atualizada)
```

```python
# A ORIGINAL CONTINUA INTACTA
print(tupla_original)
```

### Iteração em tuplas

Iteramos da mesma forma que em listas.

## 5. Strings como sequências imutáveis

Em Python, textos são representados pelo tipo `str`. Uma string é uma sequência ordenada de caracteres e compartilha o mesmo sistema de indexação e fatiamento que já vimos em listas e tuplas.

```python
curso = "Estatística"

print("Primeiro caractere:", curso[0])
print("Último caractere:", curso[-1])
print("Comprimento da palavra:", len(curso))
```

```python
# Fatiamento em strings
print("Quatro primeiros caracteres:", curso[:4])
print("Do índice 4 ao fim:", curso[4:])
print("Pulando de 2 em 2:", curso[::2])
print("Texto invertido:", curso[::-1])
```

Assim como as tuplas, as strings são imutáveis. Isso significa que você nunca consegue trocar uma letra isolada dentro de uma string já criada.

Qualquer operação que modifique um texto na verdade cria e retorna uma nova string na memória.

```python
mensagem = "gato"
mensagem[0] = "r"
```

```python
# Para obter "rato", precisamos gerar uma nova string combinando pedaços
nova_mensagem = "r" + mensagem[1:]
print("Mensagem original mantida:", mensagem)
print("Nova mensagem criada:", nova_mensagem)
```

Strings também suportam iteração direta com o laço `for` e verificação de subtextos com o operador `in`.

```python
palavra = "Python"

for letra in palavra:
    print("Caractere:", letra)

print("th" in palavra)
print("z" in palavra)
```

## 6. Métodos comuns de transformação de strings

O Python oferece métodos nativos muito poderosos para manipular textos.

Ajuste de maiúsculas e minúsculas:
- `lower()`: converte todos os caracteres para minúsculas.
- `upper()`: converte todos os caracteres para maiúsculas.
- `capitalize()`: primeira letra da frase em maiúscula, restante em minúscula.
- `title()`: primeira letra de cada palavra em maiúscula.

```python
texto = "linguagem de programação PYTHON"

print("lower():", texto.lower())
print("upper():", texto.upper())
print("capitalize():", texto.capitalize())
print("title():", texto.title())
```

Remoção de espaços em branco e quebras de linha:
- `strip()`: remove espaços extras no início e no final do texto.
- `lstrip()`: remove espaços apenas no início (lado esquerdo).
- `rstrip()`: remove espaços apenas no final (lado direito).

```python
entrada_usuario = "    ana.silva@email.com   \n"

# repr(...) usado para melhor visualizar os caracteres invisíveis (espaço ' ' e quebra de linha '\n')
print("Texto com espaços indesejados:", repr(entrada_usuario))

limpo_strip = entrada_usuario.strip()
limpo_lstrip = entrada_usuario.lstrip()
limpo_rstrip = entrada_usuario.rstrip()

print("Texto limpo com strip():", repr(limpo_strip))
print("Texto limpo com lstrip():", repr(limpo_lstrip))
print("Texto limpo com rstrip():", repr(limpo_rstrip))
```

Busca e substituição de trechos:
- `find(subtexto)`: retorna a primeira posição onde o subtexto aparece (ou -1 se não encontrar).
- `count(subtexto)`: conta as ocorrências do subtexto.
- `replace(antigo, novo)`: substitui ocorrências de um trecho por outro.

```python
frase = "O café quente esfriou, mas ainda é café."

print("Ocorrências da palavra 'café':", frase.count("café"))
print("Primeiro índice de 'café':", frase.find("café"))
print("Índice de 'chá' (não existe):", frase.find("chá"))
```

```python
frase_alterada = frase.replace("café", "chá")
print("Frase após replace:", frase_alterada)
```

Como strings são imutáveis, note que `replace` não altera a original (mas sim gera uma nova string):

```python
print(frase) # continua intacta
```

Ao usar `texto.replace(a_remover, "")`, cada ocorrência `a_remover` será removida de `texto` (pois é substituída pela string vazia `""`):

```python
print("Texto Original com várias letras o".replace("o", ""))
```

```python
print("Texto Original com várias letras o".replace("ri", ""))
```

Verificações de formato (retornam booleanos `True` ou `False`):
- `startswith(prefixo)`: confere se o texto começa com determinado trecho.
- `endswith(sufixo)`: confere se o texto termina com determinado trecho.
- `isdigit()`: confere se a string contém apenas dígitos numéricos.
- `isalpha()`: confere se a string contém apenas letras (sem espaços ou pontuação).

```python
documento = "relatorio_final.pdf"
codigo = "123456"
nome = "Carlos1"

print("É um arquivo PDF?:", documento.endswith(".pdf"))
print("Começa com 'relatorio'?:", documento.startswith("relatorio"))
print("O código possui apenas números?:", codigo.isdigit())
print("O nome possui apenas letras?:", nome.isalpha())
```

## 7. Divisão e junção de textos: conectando strings e listas

Dois dos métodos mais importantes para processamento de dados textuais são `split()` e `join()`.

- `split(separador)`: quebra uma string em vários pedaços e gera uma lista de palavras.
- `join(lista)`: junta todos os elementos de uma lista de textos em uma única string, usando um caractere de ligação.

```python
frase = "Python é simples, legível e direto"

# Quebra pelos espaços em branco por padrão
palavras = frase.split()
print("Lista de palavras:", palavras)
print("Tipo do resultado:", type(palavras))
```

```python
# Quebrando por um delimitador específico, como vírgula ou ponto e vírgula
linha_csv = "mouse;teclado;monitor;headset"
itens = linha_csv.split(";")

print("Itens separados:", itens)
```

```python
# Fazendo o caminho inverso com join(): transformando a lista de volta em texto
tecnologias = ["Python", "SQL", "Git", "Docker"]

texto_conectado = " - ".join(tecnologias)
print("Resultado com join(' - '):", texto_conectado)

texto_virgula = ", ".join(tecnologias)
print("Resultado com join(', '):", texto_virgula)
```

## Formatação

Em f-strings, podemos controlar o tamanho mínimo ocupado por um texto (para que textos de tamanhos distintos ocupem um mesmo espaço em uma tabela por exemplo). Nesse contexto, indicamos com `>`, `<`, `^` se queremos que o texto fique alinhado à direita, esquerda, ou centralizado:
- `{texto:>10}`: alinha o texto à direita ocupando no mínimo 10 espaços. (`>` pode ser omitido)
- `{texto:<10}`: alinha o texto à esquerda ocupando no mínimo 10 espaços.
- `{texto:^10}`: centraliza o texto em um espaço de 10 caracteres.

```python
# Uma tabela representada como uma LISTA de TUPLAS
dados_produtos = [
    ("Caderno", 18.5),
    ("Caneta Azul", 2.2),
    ("Diamante", 14299925.0),
    ("Régua 30cm", 4.75)
]

cabecalho = f"{'Produto':<18} | {'Valor':>9}"
N = len(cabecalho)

print("-" * N)
print(cabecalho)
print("-" * N)

for nome, valor in dados_produtos: # desempacotamento de tupla
    print(f"{nome:<18} | R$ {valor:>6.2f}") # 6 indica que o tamanho MÍNIMO é 6 caracteres

print("-" * N)
```

Acima, de propósito, o preço do diamante ultrapassa o tamanho mínimo 6. Fora isso, a tabela ficou alinhada!

### Exercício 1: análise de coordenadas com tuplas

Dado o par ordenado `ponto_a = (4, 7)` e `ponto_b = (1, 3)`:
- Desempacote os valores de cada ponto em variáveis `xa, ya` e `xb, yb`.
- Calcule a distância euclidiana entre os dois pontos dada pela fórmula:
  distancia = raiz quadrada de ((xb - xa)² + (yb - ya)²)
- Exiba a distância formatada com 2 casas decimais.

(Lembre-se: calcular a raiz quadrada equivale a elevar à potência 0.5)

```python
ponto_a = (4, 7)
ponto_b = (1, 3)

# Desempacote os pontos aqui:
xa,ya = ponto_a
xb,yb = ponto_b

# Calcule a distância euclidiana:
delta_x = xa - xb
delta_y = ya - yb
euclidiana = (delta_x**2 + delta_y**2)**0.5


# Imprima o resultado formatado:
print(round(euclidiana, 2))
```
