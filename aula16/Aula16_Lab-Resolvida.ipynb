## Exercício 3: Estatística básica

Neste exercício você implementará cálculos estatísticos essenciais aplicando laços `for`, acumuladores e indexação sobre a lista. Pode usar `sum` e `len`, se quiser.

### Fórmulas necessárias:
1. **Média ($\mu$):**
   $$\mu = \frac{\sum x_i}{N}$$
   (Soma de todos os elementos dividida pelo número total de elementos).

2. **Variância populacional ($\sigma^2$):**
   $$\sigma^2 = \frac{\sum (x_i - \mu)^2}{N}$$
   (A média dos desvios ao quadrado em relação à média geral).

3. **Desvio padrão ($\sigma$):**
   $$\sigma = \sqrt{\sigma^2}$$
   (A raiz quadrada da variância. Em Python, você pode obter a raiz elevando a $0.5$, ou seja, `valor ** 0.5`).

```python
notas = [6.5, 8.0, 7.5, 4.0, 9.5, 5.0, 8.5]

# Calcule abaixo (pode definir novas variáveis, usar laços, etc.!)

media = sum(notas) / len(notas)

variancia = 0
summ = 0
for n in notas:
    summ += (n - media)**2
variancia = summ / len(notas)

desvio_padrao = variancia**0.5


print(f'Média: {media:.2f} | Variância: {variancia:.2f} | Desvio padrão: {desvio_padrao:.2f}')
```

Faça um programa para responder à seguinte pergunta:
- Quantos alunos estão a menos de 1 desvio padrão da média?

```python
notas = [6.5, 8.0, 7.5, 4.0, 9.5, 5.0, 8.5]

# Calcule abaixo (pode definir novas variáveis, usar laços, etc.!)
# Pode reusar seu cálculo de média e desvio padrão obtido na célula anterior

quantidade_alunos_perto_da_media = 0
for n in notas:
    if (abs(media - n) < desvio_padrao): 
        quantidade_alunos_perto_da_media+=1

print(f'Quantidade de alunos a menos de 1 desvio padrão da média: {quantidade_alunos_perto_da_media}')
```

## Exercício 4: Separação de valores pares e ímpares

Dada a lista inicial de números inteiros:
`valores = [14, 27, 33, 42, 50, 19, 8, 91, 104, 63, 72]`

Escreva um programa que
1. Percorra a lista `valores` e separe os números em duas novas listas: `pares` e `impares`.
2. Calcule e exiba a média dos números pares e a média dos números ímpares separadamente.
3. Identifique e informe qual das duas listas obteve a maior média.

```python
valores = [14, 27, 33, 42, 50, 19, 8, 91, 104, 63, 72]

# Escreva aqui

pares = []
impares = []
for v in valores:
    if (v % 2 == 0):
        pares.append(v)
    else:
        impares.append(v)
print(pares, impares)

media_pares = sum(pares)/len(pares)
media_impares = sum(impares)/len(impares)
print(f'Média pares: {media_pares:.2f}\nMédia ímpares: {media_impares:.2f}')


print(f"Maior: {'Pares' if media_pares >= media_impares else 'Ímpares'}")
```

## Exercício 5: Normalização min-max

Na preparação de dados para aprendizado de máquina, é muito comum transformar uma lista de números para que todos fiquem contidos no intervalo de $0.0$ a $1.0$. Essa técnica se chama **normalização min-max**.

A fórmula para transformar cada elemento $x$ da lista é
$$x_{\mathrm{normalizado}} = \frac{x - x_{\mathrm{min}}}{x_{\mathrm{max}} - x_{\mathrm{min}}}$$

Onde
* $x_{\mathrm{min}}$ é o menor valor da lista.
* $x_{\mathrm{max}}$ é o maior valor da lista.

---
Dada a lista de salários brutos:
`salarios = [1500.0, 2300.0, 4200.0, 1800.0, 9500.0, 3100.0]`

Escreva um programa que
1. Construa e exiba uma nova lista chamada `salarios_normalizados` aplicando a fórmula em cada valor.

```python
salarios = [1500.0, 2300.0, 4200.0, 1800.0, 9500.0, 3100.0]

# Escreva aqui

minimum = min(salarios)
maximum = max(salarios)

salarios_normalizados = []
for s in sorted(salarios):
    norm = (s - minimum) / (maximum - minimum)
    salarios_normalizados.append(norm)

print(salarios_normalizados)
```

## Exercício 6: Eliminação de elementos duplicados

Dada a lista de identificadores de acesso a um sistema:
`acessos = [102, 105, 101, 102, 108, 105, 102, 110, 108, 101, 115]`

Escreva um programa que
1. Crie uma nova lista chamada `primeiros_acessos` que contenha apenas a primeira aparição de cada identificador, preservando a ordem em que apareceram originalmente.
2. Calcule e informe quantos registros duplicados foram removidos no total (ou seja, a diferença de tamanho entre a lista original e a lista de valores únicos).


```python
acessos = [102, 105, 101, 102, 108, 105, 102, 110, 108, 101, 115]
primeiros_acessos = []

acc = []
for a in acessos:
    if a in acc:
        continue
    acc.append(a)
    primeiros_acessos.append(a)

print(primeiros_acessos)
print(f'Total duplicatas removidas: {len(acessos) - len(primeiros_acessos)}')
```
