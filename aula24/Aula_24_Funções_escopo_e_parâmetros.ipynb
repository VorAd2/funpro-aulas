## 1. Escopo de variáveis: locais versus globais

O **escopo** de uma variável define em quais partes do programa ela pode ser acessada ou modificada, bem como o seu tempo de vida na memória do computador.

Em Python, existem dois níveis principais de escopo ao trabalhar com funções:
* **Escopo local:** variáveis criadas dentro de uma função. Elas só existem enquanto a função estiver sendo executada.
* **Escopo global:** variáveis criadas fora de qualquer função (no corpo principal do programa). Elas podem ser lidas a partir de qualquer ponto do código.

### Variáveis locais
Qualquer variável criada dentro de uma função é local para aquela função. Assim que a função termina sua execução, essas variáveis deixam de existir na memória.

```python
def configurar_sistema():
    versao = "2.4.1" # variável local
    print(f"Versão dentro da função: {versao}")

configurar_sistema()
```

```python
print(versao) # dá erro
```

### Variáveis globais
Variáveis definidas no fluxo principal do arquivo são globais. Funções conseguem **ler** seus valores normalmente.

```python
nome_instituicao = "UFC" # variável global

def exibir_instituicao():
    print(f"Instituição de ensino: {nome_instituicao}")

exibir_instituicao()
print(f"Acesso fora da função: {nome_instituicao}")
```

### Sombreamento de variáveis (*shadowing*)
Se criarmos uma variável local com o **mesmo nome** de uma variável global, a função dará prioridade à versão local. A variável local "esconde" (faz sombra) sobre a variável global sem alterá-la.

```python
saldo = 1000 # global

def alterar_saldo_local():
    saldo = 500 # cria uma NOVA variável local chamada 'saldo'
    print(f"Saldo dentro da função: {saldo}")

alterar_saldo_local()
print(f"Saldo fora da função (inalterado): {saldo}")
```

Por padrão, toda variável alvo de uma atribuição no escopo de uma função é marcada como **local**.

```python
teste = 0

def f_teste():
    a = teste
    teste = 2 # nome 'teste' é marcado como LOCAL, por isso a linha acima dá erro (a variável LOCAL não existia ainda)

f_teste()
```

### Modificando variáveis globais
Se você tentar reatribuir um valor para uma variável global dentro de uma função sem aviso prévio, o Python entenderá que você está criando uma variável local.

Para modificar explicitamente o valor de uma variável global de dentro de uma função, utilizamos o comando **`global`**.

```python
pontuacao = 0 # global

def adicionar_pontos():
    global pontuacao
    pontuacao = pontuacao + 10

adicionar_pontos()
adicionar_pontos()
print(f"Pontuação final atualizada: {pontuacao}")
```

> EVITE usar a instrução `global`. Quase sempre é desnecessário. Funções que alteram variáveis externas diretamente tornam o código imprevisível e difícil de debugar.

> O caminho recomendado é passar a informação por parâmetro e receber a alteração via `return`.

## 2. Parâmetros com valores padrão (*default arguments*)

Em muitas situações, uma função tem parâmetros que costumam assumir um mesmo valor na maioria dos casos.

Python permite definir **valores padrão** para esses parâmetros na própria declaração (`def`). Dessa forma, quem chama a função pode omitir esses argumentos caso deseje usar o comportamento padrão.

### Tornando parâmetros opcionais
Para definir um valor padrão, utilizamos o operador `=` diretamente no cabeçalho da função.

```python
# 'saudacao' tem como valor padrão a string "Olá"
def saudar(nome, saudacao="Olá"):
    print(f"{saudacao}, {nome}!")
```

```python
# Chamada sem informar a saudação: o Python usa o valor padrão "Olá"
saudar("Alice")
```

```python
# Chamada informando uma saudação personalizada: substitui o valor padrão
saudar("Guilherme", "Bom dia")
```

```python
def elevar_potencia(base, expoente=2):
    return base ** expoente
```

```python
# Chamada usando o padrão (calcula 5 ao quadrado)
print("5 ao quadrado:", elevar_potencia(5))
```

```python
# Chamada substituindo o padrão (calcula 2 elevado a 3)
print("2 elevado ao cubo:", elevar_potencia(2, 3))
```

> A ordem dos parâmetros importa!

Parâmetros **obrigatórios** (sem valor padrão) devem vir sempre **antes** dos parâmetros opcionais (com valor padrão).

Se você tentar colocar um parâmetro obrigatório depois de um opcional, o interpretador gerará um erro de sintaxe (`SyntaxError`).

```python
def funcao_invalida(opcao="sim", valor):
    print(f"Opção: {opcao} | Valor: {valor}")
```

> Na declaração da função, coloque primeiro todos os parâmetros sem valor padrão e, apenas no final, todos os parâmetros que possuem valor padrão.

## 3. Empacotamento de argumentos posicionais (*args)

Até agora, nossas funções sempre esperavam uma quantidade pré-definida de argumentos. Mas como criar funções que aceitam uma quantidade variável de valores (como a própria função `print()`, que aceita quantos argumentos quisermos)?

Para isso, o Python oferece o operador de empacotamento posicional: o asterisco **`*`**.

### O que é o *args?
Ao colocar um asterisco antes do nome de um parâmetro (por convenção chamado de `*args`), o interpretador reúne todos os argumentos posicionais extras em uma **tupla**.

```python
# A função recebe qualquer quantidade de argumentos posicionais
def demonstrar_args(*args):
    print("Conteúdo recebido:", args)
    print("Tipo do dado:", type(args))
```

```python
# Testando com 0, 1 e 4 argumentos
demonstrar_args()
demonstrar_args(10)
demonstrar_args("Python", True, 3.14, 42)
```

Por exemplo, podemos percorrer essa tupla usando um laço de repetição ou utilizar funções como `sum()` e `len()` diretamente sobre ela.

```python
def somar_todos(*numeros):
    total = 0
    for num in numeros:
        total += num
    return total
```

```python
# Chamando a função com diferentes quantidades de valores
print("Soma de 2 números:", somar_todos(10, 20))
print("Soma de 5 números:", somar_todos(1, 2, 3, 4, 5))
print("Soma sem argumentos:", somar_todos())
```

### Combinando parâmetros obrigatórios com *args
Podemos definir parâmetros normais antes de `*args`. Nesse caso, os primeiros argumentos são associados aos parâmetros fixos e todo o restante vai para a tupla.

```python
def registrar_estudante(nome, matricula, *notas):
    print(f"Estudante: {nome} (Matrícula: {matricula})")
    print(f"Histórico de notas: {notas}")
    if len(notas) > 0:
        media = sum(notas) / len(notas)
        print(f"Média apurada: {media:.1f}")
```

```python
# Chamando com 3 notas
registrar_estudante("Carlos", 12345, 8.0, 9.5, 7.0)
```

```python
# Chamando sem nenhuma nota lançada
registrar_estudante("Beatriz", 54321)
```

### Passando uma coleção diretamente

Tudo o que fizemos com `*args` pode ser feito criando uma função comum que recebe uma lista ou tupla como parâmetro único.

```python
# Abordagem tradicional: a função recebe uma única lista como parâmetro
def somar_colecao(numeros):
    total = 0
    for num in numeros:
        total += num
    return total
```

```python
# Ao chamar, quem usa a função é obrigado a empacotar os dados em colchetes []
lista_valores = [10, 20, 30, 40]
print("Resultado passando uma lista:", somar_colecao(lista_valores))
```

```python
# Passando uma lista criada diretamente na chamada
print("Resultado passando lista literal:", somar_colecao([5, 15, 25]))
```

Veja a diferença na chamada:

1. **Passando uma lista:** `somar_colecao([1, 2, 3])`  
   * Exige que quem chama a função coloque colchetes `[]`.
   * É ideal quando você já tem uma lista pronta no seu programa.

2. **Usando `*args`:** `somar_todos(1, 2, 3)`  
   * Os valores são passados soltos, separados por vírgula.
   * É uma conveniência sintática (*syntactic sugar*), deixando a chamada da função mais limpa (exatamente como funciona o `print(1, 2, 3)`).

> O `*args` existe para dar praticidade a quem chama a função, evitando a necessidade de digitar colchetes.

> Se você já tem a lista na memória, é mais conveniente passar a lista diretamente.


```python
def somas_separadas(A, B):
    return sum(A), sum(B)
```

```python
somas_separadas([1, 2], [3, 4])
```

Faz pouco sentido passar as listas por *args

```python
def somas_separadas_com_args(A, *B): # MENOS CONVENIENTE
    return sum(A), sum(B)
```

```python
somas_separadas_com_args([1, 2], 3, 4) # isso é MENOS conveniente, pois fica inconsistente
```

```python
somas_separadas_com_args([1, 2], [3, 4]) # por conta do asterisco, B recebe ([3, 4],): é uma tupla de 1 elemento (que é a lista [3, 4])
```

## 4. Empacotamento de argumentos nomeados (**kwargs)

Assim como o `*args` empacota valores soltos em uma tupla, o Python permite capturar múltiplos argumentos nomeados (`chave=valor`) passados na chamada de uma função.

Para isso, utilizamos o operador de dois asteriscos: **`**`**.

### O que é o **kwargs?
Ao colocar dois asteriscos antes do nome de um parâmetro (por convenção chamado de `**kwargs`, de *keyword arguments*), o interpretador reúne todos os argumentos nomeados extras em um **dicionário**.

```python
# A função recebe argumentos no formato nome_do_parametro=valor
def demonstrar_kwargs(**kwargs):
    print("Conteúdo recebido:", kwargs)
    print("Tipo do dado:", type(kwargs))
```

```python
# Chamada passando pares nome-valor variados
demonstrar_kwargs(nome="Carlos", idade=20, matricula=45678)
```

### Iterando sobre os argumentos recebidos
Como `kwargs` é um dicionário comum do Python, podemos utilizar todos os métodos que já conhecemos, como `.items()`, `.keys()` ou `.get()`.

```python
def exibir_ficha_cadastral(nome, **detalhes):
    print(f"Cadastro de: {nome}")
    print("Dados adicionais:")
    for chave, valor in detalhes.items():
        print(f" - {chave}: {valor}")
```

```python
# Chamada com dados flexíveis
exibir_ficha_cadastral("Mariana", cidade="Sobral", curso="Engenharia", semestre=3)
```

### Precisamos mesmo de **kwargs?
Assim como no caso do `*args`, você **não é obrigado** a usar `**kwargs`. Podemos simplesmente criar uma função que recebe um dicionário explícito como parâmetro.

```python
# Abordagem tradicional: recebendo um dicionário comum
def exibir_ficha_dicionario(nome, dados_adicionais):
    print(f"Cadastro de: {nome}")
    for chave, valor in dados_adicionais.items():
        print(f" - {chave}: {valor}")
```

```python
# Chamando a função passando um dicionário explícito com chaves {}
dados_aluno = {"cidade": "Sobral", "curso": "Engenharia", "semestre": 3}
exibir_ficha_dicionario("Mariana", dados_aluno)
```

### Comparando as abordagens:

1. **Passando um dicionário:** `funcao({"chave": valor})`  
   * É ideal quando os dados já estão estruturados em um dicionário no seu código.

2. **Usando `**kwargs`:** `funcao(chave=valor)`  
   * Evita a digitação das chaves `{}`, das aspas nos nomes dos campos e dos dois-pontos `:`, tornando a sintaxe de chamada mais simples e legível.

> **Resumo dos empacotadores:**
> * `*args` transforma argumentos soltos posicionais em uma **tupla** `( )`.
> * `**kwargs` transforma argumentos nomeados `chave=valor` em um **dicionário** `{ }`.

### Exercício 2: Resolução de escopo

O código a seguir tenta incrementar um contador global, mas resulta em erro.

```python
tentativas = 0

def registrar_tentativa():
    # Apesar de 'tentativas' existir globalmente, esse nome foi usado localmente, em algum momento do bloco da função,
    #como alvo de uma atribuição e, portanto, foi marcado como um recurso local. Sendo assim,
    #tentar usar 'tentativas' antes do momento de atribuição local gera erro, pois tal variável
    #passa a ser tratada como local pelo Python.
    print(f"Tentativas antes: {tentativas}")
    tentativas = tentativas + 1
    print(f"Tentativas depois: {tentativas}")

registrar_tentativa()

```

1. Execute o código e identifique qual erro ocorre.
2. Certifique-se de entender por que este erro ocorre mesmo existindo `tentativas = 0` fora da função.
3. Como corrigir o erro mantendo a ideia de atualizar a variável externa? Apresente duas formas:
   * **Estratégia A:** Usando `global`.
   * **Estratégia B (Recomendada):** Sem usar `global`, passando o valor como parâmetro e retornando o novo valor (`return`).

```python
# A
tentativas = 0

def registrar_tentativa():
    global tentativas
    print(f"Tentativas antes: {tentativas}")
    tentativas = tentativas + 1
    print(f"Tentativas depois: {tentativas}")

registrar_tentativa()


# B
tentativas = 0

def registrar_tentativa(tent):
    print(f"Tentativas antes: {tent}")
    print(f"Tentativas depois: {tent+1}")
    return tent+1

tentativas = registrar_tentativa(tentativas)
```

### Exercício 3: Eliminando o `global`

O seguinte sistema de carrinho de compras foi escrito usando variáveis globais:

```python
total_carrinho = 0.0

def adicionar_item(preco, quantidade):
    global total_carrinho
    total_carrinho += preco * quantidade

def aplicar_cupom(porcentagem):
    global total_carrinho
    total_carrinho -= total_carrinho * (porcentagem / 100)

adicionar_item(50.0, 2)   # total vai para 100.0
adicionar_item(25.0, 4)   # total vai para 200.0
aplicar_cupom(10)         # total vai para 180.0
print(f"Total: R$ {total_carrinho:.2f}")
```

Reescreva as funções `adicionar_item` e `aplicar_cupom` de modo que
- Nenhuma delas utilize a palavra-chave `global`.
- Ambas recebam o total atual como parâmetro e retornem o novo total calculado.
- O fluxo principal atualize a variável `total_carrinho` recebendo o retorno dessas funções.

```python
total_carrinho = 0.0

def adicionar_item(total_carrinho, preco, quantidade):
    return total_carrinho + (preco * quantidade)

def aplicar_cupom(total_carrinho, porcentagem):
    desc = total_carrinho * (porcentagem/100)
    return total_carrinho - desc

total_carrinho = adicionar_item(total_carrinho, 50.0, 2)   # total vai para 100.0
total_carrinho = adicionar_item(total_carrinho, 25.0, 4)   # total vai para 200.0
total_carrinho = aplicar_cupom(total_carrinho, 10)         # total vai para 180.0
print(f"Total: R$ {total_carrinho:.2f}")
```
