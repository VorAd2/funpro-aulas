# Aula 23: Modularização: funções

## 1. Por que precisamos de funções?

Até agora, nossos programas foram escritos de forma linear: comandos executados um após o outro, com desvios condicionais (`if/else`) e repetições (`for/while`).

Conforme os programas crescem, um problema surge:
-  **Repetição de código:** trechos idênticos ou muito parecidos são reescritos em vários pontos.

Isso traz dificuldade de manutenção, pois para alterar uma regra de negócio, você terá que alterar todos os lugares onde aquele código foi copiado.

### O Princípio D.R.Y. (*Don't Repeat Yourself*)
A **modularização** é a prática de dividir um programa em blocos menores, independentes e reutilizáveis. Em Python, a unidade fundamental de modularização é a **função**.

### Exemplo: código duplicado sem funções
Imagine que precisamos processar as notas de três estudantes, calculando a média e imprimindo um cabeçalho formatado para cada um.

````python
# Dados do primeiro estudante
nome1 = "Ana"
n1_ana = 8.5
n2_ana = 7.0
media_ana = (n1_ana + n2_ana) / 2

print(f"Estudante: {nome1}")
print(f"Média final: {media_ana:.1f}")
if media_ana >= 7.0:
    print("Situação: Aprovado(a)")
else:
    print("Situação: Em recuperação")
````

````python
# Dados do segundo estudante (código copiado e colado)
nome2 = "Bruno"
n1_bruno = 5.0
n2_bruno = 6.0
media_bruno = (n1_bruno + n2_bruno) / 2

print(f"Estudante: {nome2}")
print(f"Média final: {media_bruno:.1f}")
if media_bruno >= 7.0:
    print("Situação: Aprovado(a)")
else:
    print("Situação: Em recuperação")
````

````python
# Dados do terceiro estudante (mais uma repetição)
nome3 = "Carla"
n1_carla = 9.0
n2_carla = 9.5
media_carla = (n1_carla + n2_carla) / 2

print(f"Estudante: {nome3}")
print(f"Média final: {media_carla:.1f}")
if media_carla >= 7.0:
    print("Situação: Aprovado(a)")
else:
    print("Situação: Em recuperação")
````


Observe o código acima:
* Se a média de aprovação mudar de `7.0` para `6.0`, precisaremos alterar o código em 3 lugares diferentes!
* Para 100 alunos, o código teria centenas de linhas repetidas.

**Solução:** isolar essa lógica em um bloco nomeado que recebe os dados necessários e executa a tarefa. Esse bloco é uma **função**.

## 2. Declaração (`def`) e chamada de funções

Para criar uma função em Python, usamos a palavra-chave reservada **`def`** (de *define*), seguida pelo nome da função, um par de parênteses `()` e dois-pontos `:`.

### Declaração
```python
def nome_da_funcao():
    # Bloco de código indentado
    # Instruções a serem executadas

Observe a indentação acima!

### Exemplo

````python
def exibir_separador():
    print("=" * 35)
````

> Ao executar a célula acima, nada é impresso na tela!

> Definir uma função apenas ensina ao Python o que fazer **quando** a função for chamada.

Para chamar uma função, escrevemos o seu nome seguido de parênteses: `nome_da_funcao()`.

````python
# Chamando a função uma vez
exibir_separador()
````

Podemos chamar a função quantas vezes forem necessárias ao longo do programa.

````python
exibir_separador()
print("              TABELA             ")
exibir_separador()
````

O corpo de uma função pode conter quantas linhas de código forem necessárias, incluindo outras funções pré-existentes como `print()`.

````python
def exibir_menu():
    print("----- MENU PRINCIPAL -----")
    print("1 - Cadastrar Estudante")
    print("2 - Listar Notas")
    print("3 - Sair")
    print("--------------------------")
````

````python
exibir_menu()
````

A ordem de execução importa no Python!

O interpretador Python lê o arquivo de cima para baixo. Você **não pode** chamar uma função antes de ela ter sido declarada, senão receberá um erro do tipo `NameError`.

````python
saudar_usuario()

def saudar_usuario():
    print("Olá! Seja bem-vindo ao sistema.")
````

## 3. Passagem de parâmetros e argumentos

Até agora, nossas funções executavam sempre exatamente a mesma tarefa quando chamadas. Para torná-las mais úteis, podemos enviar informações para dentro delas.

### Parâmetro versus argumento
Embora esses termos sejam usados no dia a dia como sinônimos, tecnicamente existe uma diferença:
* **Parâmetro** é a variável declarada entre os parênteses na definição da função (`def`).
* **Argumento** é o valor real repassado à função no momento da sua chamada.

````python
# 'nome' é o parâmetro da função
def saudar_pessoa(nome):
    print(f"Olá, {nome}! Seja bem-vindo(a).")
````

````python
# Passando o argumento "Ana"
saudar_pessoa("Ana")
````

````python
# Passando o argumento por meio de uma variável existente
usuario_conectado = "Carlos"
saudar_pessoa(usuario_conectado)
````

Uma função pode receber dois ou mais parâmetros. Para isso, separamos seus nomes por vírgula dentro dos parênteses.

````python
# Definindo uma função com três parâmetros
def exibir_relatorio_simples(nome, disciplina, media):
    print(f"Estudante: {nome}")
    print(f"Disciplina: {disciplina}")
    print(f"Média calculada: {media:.1f}")
````

**Chamada com argumentos posicionais**: a correspondência entre os argumentos e os parâmetros da função é feita puramente pela ordem em que aparecem.

````python
# Chamada com argumentos posicionais respeitando a ordem: nome, disciplina, media
exibir_relatorio_simples("Jorge", "FUP", 9.5)
````

````python
# Inverter a ordem gera problemas!
exibir_relatorio_simples(9.5, "Algoritmos", "Beatriz")
````

**Chamada com argumentos nomeados:** podemos informar explicitamente o nome de cada parâmetro na chamada da função. Nesse caso, a ordem em que passamos os valores não importa.

````python
# Chamada com argumentos nomeados fora de ordem
exibir_relatorio_simples(disciplina="FUP", media=10, nome="Nelma")
````

Se uma função exige parâmetros e não fornecemos todos os argumentos necessários, o Python interrompe a execução com um erro do tipo `TypeError`.

````python
saudar_pessoa()
````

## 4. O comando return e valores de retorno

Até agora, todas as nossas funções apenas imprimiam mensagens na tela usando `print()`. Em muitos livros, funções assim são chamadas de **procedimentos**.

Frequentemente queremos que uma função **retorne uma resposta** para quem a chamou.

Para enviar uma resposta de volta ao programa principal, utilizamos a palavra reservada **`return`**.

> Alguns autores reservam o termo "função" especificamente para esse tipo de bloco que retorna algum valor.

````python
def calcular_media(nota1, nota2):
    return (nota1 + nota2) / 2
````

````python
media = calcular_media(7, 9)
print(media)
````

### print() e return
Veja que
* **`print()`** apenas exibe um texto visual no terminal. Esse valor não pode ser reaproveitado pelo programa para outros cálculos.
* **`return`** devolve o resultado diretamente para a linha onde a função foi chamada, permitindo guardar o valor em variáveis ou utilizá-lo em novas expressões.

````python
def exibir_media(nota1, nota2):
    print((nota1 + nota2) / 2)
````

````python
media = exibir_media(7, 9)
print(media) # media NÃO CARREGA A MÉDIA, mas sim a sentinela None
````

````python
media_mais_1 = media + 1 # erro, pois media não é número
````

Ocorreram 2 prints acima: 1 quando `exibir_media` foi chamado (exibiu 8.0), e outro pelo `print(media)` (exibiu None).

> Quando uma função não possui a palavra-chave `return`, o interpretador Python faz com que ela retorne um valor especial nulo chamado **`None`**.

````python
def exibir_media(nota1, nota2):
    print((nota1 + nota2) / 2)
    return None # <-- inserido automaticamente, se omitido
````

Como a função devolve um número real (`float`), podemos usá-la diretamente dentro de estruturas de controle (`if/else`) ou em outras expressões aritméticas.

````python
# Usando o retorno diretamente em uma condicional
if calcular_media(5.0, 6.0) >= 7.0:
    print("Estudante aprovado(a) direto!")
else:
    print("Estudante precisará fazer prova final.")
````

### Encerramento com `return`

Assim que o interpretador executa o comando`return`, a execução da função é **interrompida imediatamente**.

````python
def teste_interrupcao():
    print("Esta linha será executada.")
    return "Fim da função"
    print("Esta linha NUNCA será exibida!")

mensagem = teste_interrupcao()
print(mensagem)
````

## 5. Retorno antecipado e múltiplos valores de retorno

O comando `return` oferece grande flexibilidade no controle de fluxo de um programa. Podemos usá-lo em diferentes ramificações condicionais para interromper a execução assim que obtivermos uma resposta, além de podermos devolver mais de um dado ao mesmo tempo.

> Podemos ter mais de uma declaração `return` dentro de uma mesma função. A primeira declaração `return` atingida encerra imediatamente o processamento da função.

````python
# Função que avalia a situação escolar e retorna uma string
def classificar_desempenho(media):
    if media >= 7.0:
        return "Aprovado(a)"
    elif media >= 4.0:
        return "Em recuperação"
    else:
        return "Reprovado(a)"
````

````python
print("Nota 8.5:", classificar_desempenho(8.5))
````

````python
print("Nota 5.5:", classificar_desempenho(5.5))
````

````python
print("Nota 2.0:", classificar_desempenho(2.0))
````

### Retorno de múltiplos valores simultâneos
Diferente de muitas linguagens de programação, o Python permite retornar dois ou mais valores separados por vírgula em uma única linha.

> Na realidade, o interpretador empacota esses valores em uma **tupla**.

> Então apenas 1 valor é retornado: a tupla! (e quem recebe a tupla tem acesso a vários valores)

````python
# Função que calcula e devolve o menor e o maior elemento de uma lista
def obter_limites(numeros):
    menor = min(numeros)
    maior = max(numeros)
    return menor, maior
````

````python
valores = [15, 42, 8, 99, 23]
resultado_tupla = obter_limites(valores)

print("Resultado empacotado:", resultado_tupla)
print("Tipo do dado retornado:", type(resultado_tupla))
````

````python
menor_valor, maior_valor = obter_limites(valores) # desempacotando a tupla

print(f"O menor valor encontrado foi: {menor_valor}")
print(f"O maior valor encontrado foi: {maior_valor}")
````

### Exercício 1

Crie uma função chamada `calcular_retangulo(base, altura)` que receba como parâmetros o comprimento da base e da altura de um retângulo. A função deve calcular e retornar dois valores: a área ($base \times altura$) e o perímetro ($2 \times (base + altura)$).

````python
# Defina calcular_retangulo aqui
def calcular_retangulo(base, altura):
    return base*altura, (base + altura)*2

area_calculada, perimetro_calculado = calcular_retangulo(5.0, 3.0)

print(f"Área do retângulo: {area_calculada} m²")
print(f"Perímetro do retângulo: {perimetro_calculado} m")
````

### Exercício 2

Crie uma função chamada `eh_par(numero)` que receba um número inteiro e retorne o valor booleano `True` se o número for par, ou `False` se for ímpar. Em seguida, utilize essa função para filtrar apenas os números pares de uma lista predefinida.

````python
valores_amostra = [12, 7, 19, 24, 30, 41, 56]
numeros_pares = []

# Defina eh_par aqui
def eh_par(numb):
    return (numb % 2 == 0)

# Itere sobre a lista aqui
for n in valores_amostra:
    if eh_par(n):
        numeros_pares.append(n)

print("Lista original:", valores_amostra)
print("Apenas os pares:", numeros_pares)
````

### Exercício 3

Crie uma função chamada `analisar_notas(lista_notas)` que receba uma lista com notas de avaliações. A função deve calcular e retornar três dados: a média da turma, a maior nota e a menor nota. Pode utilizar as funções nativas do Python (`sum`, `len`, `max`, `min`) para simplificar o cálculo.

````python
notas_semestre = [7.5, 8.0, 4.5, 9.5, 6.0, 10.0, 5.5]

# Defina analisar_notas aqui
def analisar_notas(lista):
    return sum(lista)/len(lista), max(lista), min(lista)

media_turma, maior_nota, menor_nota = analisar_notas(notas_semestre)
print(f"Média geral: {media_turma:.2f}")
print(f"Maior nota registrada: {maior_nota:.1f}")
print(f"Menor nota registrada: {menor_nota:.1f}")
````
