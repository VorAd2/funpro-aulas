## 1. De array ao tipo Series

No NumPy, os arrays são acessados por posições numéricas inteiras (0, 1, 2...).

O `Series` da biblioteca Pandas é uma estrutura unidimensional que une o comportamento de um array NumPy (vetorização e homogeneidade de tipo) à semântica de um dicionário de Python (indexado por chaves imutáveis quaisquer).

```python
import pandas as pd
import numpy as np
```

### Criando uma Series a partir de uma lista

Considere medições de temperatura corporal (em °C) de uma amostra de pacientes em triagem hospitalar.

```python
temperaturas = [36.5, 37.2, 38.6, 36.8]
serie_temp = pd.Series(temperaturas)
```

```python
print(serie_temp)
```

### Índice e dados em Series

A coluna da esquerda representa o índice (`index`). A coluna da direita contém os valores (`values`), que são armazenados internamente como um array NumPy.

```python
print("Valores (array subjacente):", serie_temp.values)
```

```python
print("Tipo da sequência:", type(serie_temp.values))
```

```python
print("Tipo do dado armazenado:", serie_temp.dtype)
print("Acessando o tipo de dado pela array subjacente:", serie_temp.values.dtype)
```

```python
print("Índice padrão:", serie_temp.index)
```

```python
for x in serie_temp.index:
    print(x)
```

```python
for x in serie_temp: # iterar no Series significa iterar nos seus valores (values)
    print(x)
```

### Definindo rótulos explícitos no índice

Diferente de uma lista comum, podemos usar texto como índice de uma linha.

```python
rotulos_pacientes = ["Carla", "Bob", "Ana", "Diana"]
serie_rotulada = pd.Series(temperaturas, index=rotulos_pacientes, name="temperatura")
```

```python
print(serie_rotulada)
```

### Criação a partir de um dicionário

O Pandas aceita um dicionário diretamente: as chaves tornam-se o índice e os valores preenchem os dados da série.

```python
dados_dict = {"Carla": 36.5, "Bob": 37.2, "Ana": 38.6, "Diana": 36.8}
serie_de_dict = pd.Series(dados_dict)
```

```python
print(serie_de_dict)
```

### Acesso por chave/rótulo

Podemos recuperar um valor pelo rótulo com a mesma sintaxe de acesso a chaves de dicionários.

```python
temp_paciente_Ana = serie_rotulada["Ana"]
print("Temperatura da Ana:", temp_paciente_Ana)
```

### Operações vetorizadas na Series

Como a Series herda as propriedades do NumPy, todas as transformações matemáticas continuam vetorizadas e preservam os rótulos associados.

```python
# Convertendo todas as observações para a escala Fahrenheit
serie_f = (serie_rotulada * 9 / 5) + 32
print(serie_f)
```

## 2. A estrutura DataFrame

Em um array NumPy bidimensional (aula 31), as células da estrutura 2D precisavam ter o mesmo tipo.

Frequentemente lidamos com cenários em que cada coluna representa uma variável diferente: algumas numéricas contínuas (`float`), outras discretas (`int`) ou categóricas (`str`/`object`).

O `DataFrame` do Pandas resolve isso funcionando como um conjunto tabular de várias `Series` alinhadas pelo mesmo índice (séries paralelas).

### Criação a partir de um dicionário de listas

## 3. Seleção e indexação de dados: colchetes, loc e iloc

Em Python usamos chaves para dicionários (`dict[chave]`) e índices inteiros para listas (`lista[i]`). No Pandas, o DataFrame unifica essas abordagens com operadores específicos para evitar ambiguidades.

### Selecionando uma única coluna

Usamos a sintaxe de dicionário. O retorno de uma coluna isolada é uma `Series`.

```python
serie_pressao = df["pressao_sistolica"]
```

```python
print(serie_pressao)
```

```python
print(type(serie_pressao))
```

### Selecionando múltiplas colunas

Passamos uma lista de strings dentro dos colchetes (`df[[...]]`). O retorno é um novo `DataFrame`.

```python
variaveis_interesse = ["idade", "pressao_sistolica"]
sub_tabela = df[variaveis_interesse]
```

```python
print(sub_tabela)
```

### Indexação posicional por inteiros: iloc

O método `iloc` (integer-location) funciona como a indexação bidimensional do NumPy (`[linha, coluna]`), ignorando qualquer rótulo de texto.

```python
df
```

```python
# Primeira linha completa (linha de índice 0)
primeiro_registro = df.iloc[0]
print(primeiro_registro)
```

```python
# Elemento específico: linha 1, coluna 2 (pressão sistólica do segundo indivíduo)
valor_especifico = df.iloc[1, 2]
print("Valor posicional [1, 2]:", valor_especifico)
```

```python
# Fatiamento posicional (como no numpy)
df.iloc[0:3, 0:2] # Linhas 0, 1, 2; Colunas 0, 1
```

### Indexação rotulada: loc

O método `loc` (location) pesquisa diretamente pelos nomes textuais das linhas e das colunas.

```python
df
```

Consulta pontual por rótulo de linha e coluna:

```python
pressao_ana = df.loc["Ana", "pressao_sistolica"]
print("Pressão da paciente Ana:", pressao_ana)
```

Ao usar `loc` com texto, o intervalo de fatiamento é **inclusivo** em ambas as pontas (o elemento final é mantido).

```python
# inclui o Bob, que está entre Carla e Ana
fatia_rotulada = df.loc["Carla":"Ana", ["idade", "fumante"]]
```

```python
print(fatia_rotulada)
```

Também podemos usar listas para selecionar as linhas.

```python
fatia_rotulada_por_lista = df.loc[["Carla", "Ana"], ["idade", "fumante"]]
print(fatia_rotulada_por_lista)
```

Demonstração: onde [] funciona e onde falha

```python
# Extrair colunas
df[["idade", "sexo"]]
```

```python
# Filtro booleano trazendo todas as colunas
df[df["idade"] > 30]
```

```python
# FALHA: tentar passar [linha, coluna] simultaneamente gera erro
#df[df["idade"] > 30, "idade"] # tenta achar 1 linha de CHAVE (df["idade"] > 30, "idade"): não existe
```

```python
# Ao usar .loc, o mesmo caso anterior funciona!
df.loc[df["idade"] > 30, "idade"]
```

O .loc funciona para esse caso! Inclusive para selecionar múltiplas colunas:

```python
# loc[linhas, colunas]
df.loc[df["idade"] > 30, ["sexo", "pressao_sistolica"]]
```

Dá até para selecionar uma porção do DataFrame a ser alterada.

```python
df
```

```python
# Atualiza apenas a coluna "pressao_sistolica" dos pacientes fumantes
df.loc[df["fumante"] == True, "pressao_sistolica"] += 5.0
```

```python
df[["fumante", "pressao_sistolica"]]
```

Em resumo, para indexação normal:
- `df[string]`: seleciona coluna
- `df[lista_de_string]`: seleciona colunas
- `df[lista_de_booleanos]`: seleciona linhas
- `df[qualquer_coisa_com_vírgula]`: **ERRO** (use .loc ou .iloc)

Para indexação por .loc:
- `df.loc[rotulo_linha]`: seleciona uma linha pelo nome (retorna `Series`).
- `df.loc[lista_de_rotulos]`: seleciona várias linhas pelos nomes.
- `df.loc[lista_de_booleanos]`: filtra linhas por condição lógica.
- `df.loc[inicio:fim]`: fatia linhas por texto (**inclusivo**: seleciona também o elemento fim).
- `df.loc[rotulo_linha, rotulo_coluna]`: valor na coordenada exata (como matriz).
- `df.loc[linhas, rotulo_coluna]`: filtra linhas e seleciona uma única coluna.
- `df.loc[linhas, lista_de_colunas]`: filtra linhas e seleciona colunas específicas.
- `df.loc[qualquer_posicao_inteira]`: **ERRO** se o índice da tabela for composto por texto.

Para .iloc:
- `df.iloc[0]`: seleciona a primeira linha da tabela (retorna `Series`).
- `df.iloc[lista_de_inteiros]`: seleciona linhas pelas posições numéricas (ex: `df.iloc[[0, 2]]`).
- `df.iloc[inicio:fim]`: fatia linhas por posição (**exclusivo**: descarta o "fim", igual a `range`).
- `df.iloc[linha_int, coluna_int]`: valor na coordenada numérica exata (como matriz NumPy).
- `df.iloc[linhas, lista_de_colunas_int]`: seleciona linhas e colunas por posições numéricas.
- `df.iloc[qualquer_texto_ou_rotulo]`: **ERRO** (não aceita strings em nenhuma hipótese).

### Unnamed: 0 ?

Ao exportar com `.to_csv()`, os rótulos de linha (os índices) foram gravados como a primeira coluna do arquivo. Ao ler sem parâmetros, o Pandas gerou um índice numérico novo (0, 1, 2, 3, ...) e tratou a primeira coluna do csv (a dos nomes) como uma coluna de nome `Unnamed: 0` no DataFrame.

## 5. Filtragem condicional, novas variáveis e agrupamentos

Na programação estruturada, filtrar dados e calcular estatísticas por subgrupos exigia combinar laços `for`, estruturas `if` e dicionários de acumuladores. No Pandas, essas tarefas são resolvidas de forma declarativa e vetorizada.

### Filtragem condicional (máscaras booleanas em DataFrames)

A sintaxe de filtragem avalia uma condição lógica em uma coluna e retorna apenas as linhas onde o resultado é `True`.

```python
# Filtrando apenas pacientes com status de fumante ativo
filtro_fumantes = dados_ajustados["fumante"] == True
print(filtro_fumantes)
```

```python
dados_fumantes = dados_ajustados[filtro_fumantes]
print(dados_fumantes)
```

### Filtros compostos com operadores lógicos

Podemos usar `&` para conjunção (E) e `|` para disjunção (OU), delimitando cada teste relacional com parênteses.

```python
# Critério de risco: idade superior a 30 anos E pressão sistólica acima de 120 mmHg
condicao_risco = (dados_ajustados["idade"] > 30) & (dados_ajustados["pressao_sistolica"] > 120.0)
```

```python
grupo_risco = dados_ajustados[condicao_risco]
print(grupo_risco)
```

### Criando novas colunas (variáveis derivadas)

Similar à atribuição em dicionários (`dict[nova_chave] = valor`), basta declarar o nome da nova coluna.

```python
# Criação de um indicador booleano de alerta hipertensivo
dados_com_alerta = dados_ajustados.copy()
dados_com_alerta["alerta_hipertensao"] = dados_com_alerta["pressao_sistolica"] >= 140.0
```

```python
dados_com_alerta
```

```python
dados_com_alerta[["pressao_sistolica", "alerta_hipertensao"]]
```

### Agrupamento com groupby

Em Python sem bibliotecas, para calcular a média por categoria, precisávamos percorrer a lista alimentando um dicionário manual de somatórios e contadores.

O método `groupby` automatiza o padrão **dividir-aplicar-combinar** (*split-apply-combine*):
1. Divide os dados com base nos níveis de uma variável categórica.
2. Aplica uma função (como `mean()`, `std()` ou `count()`).
3. Combina os resultados em uma nova tabela.

```python
# Média da pressão sistólica segmentada pela condição de fumante
media_por_habito = dados_ajustados.groupby("fumante")["pressao_sistolica"].mean()
```

```python
print(media_por_habito)
```

Agrupamento com múltiplas variáveis numéricas simultâneas:

```python
# Médias de idade e pressão estratificadas por sexo
medias_por_sexo = dados_ajustados.groupby("sexo")[["idade", "pressao_sistolica"]].mean()
```

```python
print(medias_por_sexo)
```

Contagem amostral por estrato (frequência absoluta de observações por grupo):

```python
categorias, contagem = np.unique(dados_ajustados['sexo'], return_counts=True)
```

```python
print(contagem)
```

```python
pd.Series(contagem, index=categorias, name="Contagem")
```

O método `.agg()` (*aggregate*) expande o `groupby`. Em vez de calcular apenas uma métrica por vez, ele permite extrair múltiplos parâmetros descritivos em uma única passada.

Passamos uma lista de strings com os nomes das métricas estatísticas desejadas.

```python
metricas = ["mean", "std", "median", "count"]
```

```python
# Média, desvio padrão, mediana e contagem da pressão por grupo de fumantes
resumo_pressao = dados_ajustados.groupby("fumante")["pressao_sistolica"].agg(metricas)
```

```python
print(resumo_pressao)
```

Funções diferentes para variáveis diferentes (mapeamento com dicionário).

```python
# Queremos a média e desvio da pressão, mas apenas mediana e valor máximo para idade
plano_analise = {
    "pressao_sistolica": ["mean", "std"],
    "idade": ["median", "max"]
}
```

```python
resumo_estratificado = dados_ajustados.groupby("sexo").agg(plano_analise)
```

```python
print(resumo_estratificado)
```

O `.agg()` também funciona na tabela global sem `groupby`.

```python
dados_ajustados[["idade", "pressao_sistolica"]].agg(["mean", "std"])
```

### Cenário: rendimento acadêmico em diferentes cursos

Uma amostra contendo horas semanais de dedicação aos estudos e nota final obtida em um exame.

```python
dados_estudantes = {
    "aluno": ["A01", "A02", "A03", "A04", "A05", "A06", "A07", "A08"],
    "curso": ["CC", "EST", "CC", "SI", "EST", "CC", "SI", "EST"],
    "horas_estudo": [12.0, 5.5, 18.0, 8.0, 14.5, 6.0, 10.5, 16.0],
    "nota_final": [7.8, 4.2, 9.5, 6.0, 8.4, 5.0, 6.8, 8.9]
}
df_alunos = pd.DataFrame(dados_estudantes)
```

```python
df_alunos.head(4)
```

### Passo 1: Persistência e recarregamento via CSV

Gravamos a tabela em disco e a reabrimos simulando a importação de dados externos.

```python
df_alunos.to_csv("desempenho_academico.csv", index=False) # index=False: não salva o índice
```

```python
df_carregado = pd.read_csv("desempenho_academico.csv") # indexcol omitido: cria novo índice 0, 1, 2, ...
```

```python
df_carregado.info()
```

### Passo 2: Estatísticas

Inspeção das medidas de tendência central e dispersão de toda a amostra.

```python
df_carregado.describe()
```

### Passo 3: Filtragem por limiar

Extraímos apenas as observações com nota superior à média global da amostra.

```python
media_global = df_carregado["nota_final"].mean()
print("Média global:", media_global)
```

```python
filtro_acima_media = df_carregado["nota_final"] > media_global
```

```python
alunos_destaque = df_carregado.loc[filtro_acima_media, ["aluno", "curso", "nota_final"]]
print(alunos_destaque)
```

### Passo 4: Sumarização por curso

Comparamos média, desvio padrão e tamanho amostral (`count`) das notas entre os diferentes cursos.

```python
resumo_estatistico = df_carregado.groupby("curso")["nota_final"].agg(["mean", "std", "count"])
```

```python
print(resumo_estatistico)
```
