## Exercício 1: listas paralelas

Considere o seguinte uso de **listas paralelas**: duas ou mais listas em que o elemento na posição `i` de uma lista corresponde ao elemento na mesma posição `i` da outra lista:

* `estudantes = ["Alice", "Bernardo", "Clara", "Diego", "Eduarda", "Felipe"]`
* `medias = [8.5, 5.0, 9.2, 6.8, 4.5, 7.0]`

Escreva um programa que
1. Calcule a média geral da turma.
2. Identifique e exiba o nome do estudante com a maior nota (e qual foi sua nota).
3. Exiba uma lista contendo apenas os nomes dos estudantes que ficaram com média igual ou superior à média geral da turma.

```python
estudantes = ["Alice", "Bernardo", "Clara", "Diego", "Eduarda", "Felipe"]
medias = [8.5, 5.0, 9.2, 6.8, 4.5, 7.0]

# Escreva seu código abaixo

avg = sum(medias) / len(medias)
print(f'Média geral: {avg:.2f}')

largest_m = 0
index = 0
acima_media = []
for i, m in enumerate(medias):
    if m > largest_m:
        largest_m = m
        index = i
    if m >= avg:
        acima_media.append(estudantes[i])

student = estudantes[index]
print(f'Estudante de destaque: {student}, com média {largest_m}')

print(acima_media)
```

## Exercício 2: série temporal

No mercado financeiro, analistas costumam calcular a variação diária de um ativo comparando o preço do dia atual $d$ em relação ao dia anterior:
$$\Delta = P_{d} - P_{d-1}$$

Dada a sequência de preços de fechamento de uma ação ao longo de 7 dias úteis:

`fechamentos = [102.50, 104.00, 101.20, 101.20, 107.80, 105.00, 110.30]`

Escreva um programa que
1. Calcule as variações diárias e armazene os resultados em uma nova lista chamada `variacoes`. Note que essa lista terá 6 elementos (já que o primeiro dia não tem dia anterior para comparação).
2. Localize em qual transição de dia ocorreu a maior alta (maior valor positivo de variação) e exiba o valor dessa variação.
3. Conte e informe quantos dias registraram variação nula (preço inalterado em relação à véspera).

```python
fechamentos = [102.50, 104.00, 101.20, 101.20, 107.80, 105.00, 110.30]

# Escreva seu código abaixo

variacoes = []
for i in range(1, len(fechamentos)):
    d = fechamentos[i]
    d_1 = fechamentos[i-1]
    variacoes.append(d - d_1)
print(variacoes)

largest = 0.0
trans = 0
for id_1, v in enumerate(variacoes):
    if v < 0.0:
        continue
    if v > largest:
        largest = v
        trans = id_1
print(f'\nMaior alta: {largest:.2f}. Transição do dia {trans + 1} para dia {trans + 2}')

count = 0
for v in variacoes:
    if v == 0:
        count += 1
print(f'Quantidade de dias com variação nula: {count}')
```

## Exercício 4: intercalação alternada

Um sistema de distribuição de tarefas precisa revezar igualmente atividades entre dois servidores (`servidor_a` e `servidor_b`).

Dadas as duas listas de IDs de tarefas:
* `tarefas_a = [101, 103, 105, 107]`
* `tarefas_b = [202, 204, 206, 208]`

Escreva um programa que
1. Crie uma terceira lista chamada `fila_unificada`.
2. Intercale elemento a elemento das duas listas, alternando sempre um item de A seguido por um item de B, de modo que o resultado final seja:
   `[101, 202, 103, 204, 105, 206, 107, 208]`
3. Exiba a lista final resultante.



```python
tarefas_a = [101, 103, 105, 107]
tarefas_b = [202, 204, 206, 208]

fila_unificada = []

# Escreva seu código abaixo

tamanho_fila_unificada = len(tarefas_a) + len(tarefas_b)
count = 0
index = -1
for i in range(tamanho_fila_unificada):
    if count % 2 == 0:
        index += 1
    elm = tarefas_a[index] if (i % 2 == 0) else tarefas_b[index]
    fila_unificada.append(elm)
    count+=1

print(fila_unificada)
```

## Exercício 5: apuração de votos

Uma eleição para o CA utilizou uma urna eletrônica que registrou os votos em uma lista de números inteiros, onde:
* `1`: Chapa Futuro
* `2`: Chapa União
* `3`: Chapa Quente
* `0`: Voto em branco
* Qualquer outro número: Voto nulo

Dada a sequência de votos apurados:
`votos = [1, 2, 1, 1, 3, 2, 1, 0, 4, 2, 1, 3, 0, 1, 9, 2, 1, 3, 2, 2]`

Escreva um programa que
1. Conte a quantidade de votos de cada chapa (1, 2 e 3), de votos em branco (0) e de votos nulos.
2. Calcule o total de votos válidos (apenas votos nas chapas 1, 2 e 3).
3. Exiba o percentual de votos de cada chapa sobre o total de **votos válidos**.
4. Identifique e exiba o número da chapa vencedora da eleição.

```python
votos = [1, 2, 1, 1, 3, 2, 1, 0, 4, 2, 1, 3, 0, 1, 9, 2, 1, 3, 2, 2]

# Escreva seu código abaixo

counts = [0,0,0,0,0]
for v in votos:
    if v in (0,1,2,3):
        counts[v] += 1
    else:
        counts[4] += 1
print(counts)

total_validos = 0
for i in range(4):
    total_validos += counts[i]
print(f'Total de votos válidos: {total_validos}\n')

chapas = (1,2,3)
percents = []
for i in range(3):
    percent = counts[i+1]/total_validos
    percents.append(percent)
    print(f'Percentual da chapa {chapas[i]} = {percent:.2f}')

control = (0,0)
for i,percent in list(enumerate(percents)):
    chapa = i+1
    if percent > control[1]:
        control = (chapa, percent)
print(f'Chapa vencedora: {control[0]}')
    
```

## Exercício 6: controle de estoque

Uma loja de periféricos mantém o controle de inventário por meio de quatro listas correlacionadas (vetores paralelos):
* `produtos = ["Teclado", "Mouse", "Monitor", "Cabo HDMI", "Headset"]`
* `estoque_atual = [12, 4, 2, 35, 5]`
* `estoque_minimo = [10, 10, 5, 20, 10]`
* `custo_unitario = [120.0, 50.0, 750.0, 25.0, 180.0]`

Um produto está em **estado crítico** quando seu `estoque_atual` fica estritamente abaixo do `estoque_minimo`.

Escreva um programa que
1. Para cada produto crítico, calcule quantas unidades precisam ser compradas para que o estoque atinja exatamente o valor do `estoque_minimo`.
2. Exiba o nome do produto crítico, a quantidade a comprar e o subtotal de custo daquele item.
3. Ao final, informe o gasto total do pedido.

```python
produtos = ["Teclado", "Mouse", "Monitor", "Cabo HDMI", "Headset"]
estoque_atual = [12, 4, 2, 35, 5]
estoque_minimo = [10, 10, 5, 20, 10]
custo_unitario = [120.0, 50.0, 750.0, 25.0, 180.0]

# Escreva seu código abaixo

produtos_criticos = []
for i in range(5):
    atual = estoque_atual[i]
    minimo = estoque_minimo[i]
    if atual < minimo:
        produtos_criticos.append((i, minimo-atual, produtos[i]))
print(produtos_criticos)
print()

gasto_total = 0
for p in produtos_criticos:
    subtotal = p[1]*custo_unitario[p[0]]
    gasto_total += subtotal
    print(f'{p[2]} --- {p[1]} unidades --- R${subtotal}\n')
print(f'Gasto total: R${gasto_total}')

```

## Exercício 7: média móvel (janela deslizante)

Sensores frequentemente registram pequenas oscilações aleatórias chamadas de ruído. Uma técnica padrão para suavizar esses dados é a **média móvel**, calculada por meio de fatiamento (*slicing*): extrai-se uma sublista de tamanho fixo (uma "janela") e calcula-se a média desses valores.

Para uma janela de tamanho 3:
* Janela 0: elementos dos índices `0, 1, 2` $\rightarrow$ média calculada
* Janela 1: elementos dos índices `1, 2, 3` $\rightarrow$ média calculada
* Janela 2: elementos dos índices `2, 3, 4` $\rightarrow$ média calculada
* E assim por diante, até o fim da lista (nesse caso, a janela deve ter 3 elementos sempre).

Dada a lista com medições ruidosas de temperatura:
`temperaturas = [18.2, 19.0, 24.5, 18.8, 19.2, 23.9, 19.5, 19.1]`
*(Note que os picos 24.5 e 23.9 são ruídos pontuais que devem ser amenizados).*

Escreva um programa que:
1. Percorra a lista aplicando uma janela deslizante de tamanho 3 usando fatiamento.
2. Calcule a média de cada janela e adicione esse valor a uma nova lista chamada `temperaturas_suavizadas`.

(Note que uma lista de tamanho $N$ com janela 3 resulta em uma lista suavizada de tamanho $N - 2$)

```python
temperaturas = [18.2, 19.0, 24.5, 18.8, 19.2, 23.9, 19.5, 19.1]
tamanho_janela = 3

temperaturas_suavizadas = []

# Escreva seu código abaixo

for i in range(len(temperaturas)-2):
    slic = temperaturas[i:i+tamanho_janela]
    print(slic)
    avg = sum(slic) / tamanho_janela
    temperaturas_suavizadas.append(round(avg, 1))
print()
print(temperaturas_suavizadas)
```

## Exercício 8: rotação cíclica

Sistemas operacionais realizam o escalonamento de processos no modelo *round-robin*: os programas em execução ficam organizados em uma fila cíclica. Quando um ciclo termina, os processos são rotacionados sem que novos dados sejam criados.

Uma **rotação à direita de $K$ posições** move os últimos $K$ elementos da lista diretamente para o começo, deslocando os demais para frente.

Por exemplo, com a lista `[1, 2, 3, 4, 5]` e $K = 2$:
* Últimos 2 elementos: `[4, 5]`
* Elementos restantes: `[1, 2, 3]`
* Resultado após rotação: `[4, 5, 1, 2, 3]`

---
Dada a lista de identificadores de processos:
`processos = ["PID_10", "PID_20", "PID_30", "PID_40", "PID_50", "PID_60", "PID_70"]`
E um valor de deslocamento:
`k = 3`

Escreva um programa que
1. **Abordagem A (fatiamento):** gere uma nova lista chamada `rotacionada_slice` que resolva a rotação combinando duas fatias com concatenação (`+`), sem usar laços.
2. **Abordagem B (mutação no local):** pegue a lista original `processos` e aplique um laço que execute $K$ vezes a operação de remover o último elemento com `.pop()` e inseri-lo na posição zero com `.insert(0, ...)`.

Inspecione com `print` que os resultados das 2 abordagens são idênticos.

```python
processos = ["PID_10", "PID_20", "PID_30", "PID_40", "PID_50", "PID_60", "PID_70"]
k = 3

# 1. Abordagem por fatiamento (sem alterar 'processos')
rotacionada_slice = processos[len(processos)-k:] + processos[:len(processos)-k]
print(rotacionada_slice)



# 2. Abordagem por mutação no local (alterando 'processos' com pop e insert)
for i in range(k):
    elm = processos.pop()
    processos.insert(0, elm)
print(processos)
```
