# Aula 30: Programação Orientada a Objetos (POO)

## 1. Do dicionário à classe

Até a aula anterior, se precisássemos representar um aluno no sistema, poderíamos usar um dicionário:

`aluno = {"nome": "Júlia", "matricula": 12345, "notas": [8, 10]}`

Problemas dessa abordagem:
1. **Sem garantia de estrutura:** Nada impede alguém de errar a digitação de uma chave ("nomme" em vez de "nome").
2. **Dados e funções separados:** Se criarmos uma função `calcular_media(aluno)`, essa função fica solta no código, separada dos dados do aluno.

A **POO** resolve isso criando um molde formal que junta dados (características) e comportamentos (funções) sob a mesma estrutura.

* **Classe:** É um molde (ex.: uma descrição de que alunos devem ter nome, matrícula e notas; e que indica que faz sentido 'calcular média' de um aluno).
* **Objeto (Instância):** É uma ocorrência concreto de algo feito a partir do molde (ex.: uma aluna específica de nome 'Júlia', matrícula '12345' e notas `[8, 10]`).

### Criando uma classe e instanciando objetos

Usamos a palavra-chave `class`. Por convenção, geralmente o nome da classe segue o padrão **PascalCase** (iniciais maiúsculas).

```python
class Aluno:
    pass # nenhum dado nem comportamento foi descrito!
```

```python
# Criando dois objetos distintos a partir do mesmo molde
aluno1 = Aluno()
aluno2 = Aluno()

print(aluno1)
print(type(aluno1))
```

### Construtor (`__init__`) e Atributos

Um objeto sem dados e sem comportamento é inútil. Precisamos que ele guarde informações (atributos).

Para isso, declaramos uma função de nome `__init__`:
* Ela roda automaticamente toda vez que um novo objeto é criado.
* O parâmetro `self` representa **a própria instância** que está sendo criada/manipulada no momento.

Essa função é chamada de **Construtor**. No Python, indicamos o construtor pelo nome específico `__init__`

> Em outras linguagens, pode haver uma palavra reservada `constructor` ou o próprio nome da classe ser usado para indicar o construtor (ex.: função `Aluno()` dentro da classe `Aluno` é o construtor).

```python
class Aluno:
    def __init__(self, nome, matricula, curso):
        self.nome = nome
        self.matricula = matricula
        self.curso = curso
```

Passamos os valores definidos no `__init__`. O Python injeta o `self` automaticamente nos bastidores.

```python
aluno_a = Aluno("Júlia", 1001, "Estatística")
aluno_b = Aluno("César", 1002, "Ciência da Computação")
```

### Acessando atributos com a notação de ponto (`.`)
Para ler um dado guardado no objeto, usamos `nome_do_objeto.nome_do_atributo`.

```python
print(aluno_a.nome)
print(aluno_a.curso)
```

```python
print(aluno_b.nome)
print(aluno_b.matricula)
```

### Modificando atributos diretamente
Atributos podem ser reatribuídos normalmente como variáveis.

```python
aluno_a.curso = "Engenharia de Software"
print(aluno_a.curso)
```

## Métodos: Dando Comportamento aos Objetos

### Dados precisam de ações

Até aqui, nossos objetos eram apenas fichas estáticas guardando dados. Mas, no mundo real, as entidades não apenas "possuem" coisas, elas também **fazem** coisas:

* Um carro tem modelo e ano, mas ele também **acelera**, **freia** e **faz curva**.
* Uma conta de banco tem titular e saldo, mas ela também **deposita**, **saca** e **transfere**.

Um **método** é simplesmente uma função definida dentro de uma classe. A diferença para uma função comum é que o método tem acesso direto aos atributos do objeto que o chamou por meio do `self`.

### Definindo o primeiro método

O primeiro parâmetro de qualquer método dentro da classe deve ser sempre o `self`.

```python
class Aluno:
    def __init__(self, nome, matricula):
        self.nome = nome
        self.matricula = matricula

    def apresentar(self):
        print(f"Olá, meu nome é {self.nome} e minha matrícula é {self.matricula}.")
```

### Chamando métodos

A sintaxe de chamada é `objeto.metodo()`. O Python passa a própria instância para o `self` automaticamente.

```python
aluno1 = Aluno("Beatriz", 202401)
aluno2 = Aluno("Daniel", 202402)
```

```python
aluno1.apresentar()
```

```python
aluno2.apresentar()
```

### Métodos que recebem argumentos e alteram atributos

Métodos podem receber argumentos adicionais (além do `self`) e modificar os atributos internos do próprio objeto.

```python
class ContaBancaria:
    def __init__(self, titular, saldo_inicial=0):
        self.titular = titular
        self.saldo = saldo_inicial

    def depositar(self, valor):
        self.saldo += valor

    def sacar(self, valor):
        if valor <= self.saldo:
            self.saldo -= valor
            return True
        print("Saldo insuficiente!")
        return False
```

### Testando depósitos e saques

```python
conta = ContaBancaria("Fernanda", 500)
print(f"Saldo inicial: {conta.saldo}")
```

```python
conta.depositar(200)
print(f"Saldo após depósito: {conta.saldo}")
```

```python
# Tentativa de sacar mais do que o saldo disponível
conta.sacar(1000)
print(f"Saldo inalterado: {conta.saldo}")
```

## Representação em Texto (`__str__`) e Listas de Objetos

### Por que `print(objeto)` parece estranho?

Se você tentar imprimir diretamente uma instância:

`print(aluno1)`

O Python exibe algo críptico, como `<__main__.Aluno object at 0x7fa8c2...>`. Isso ocorre porque o Python não adivinha quais atributos você deseja ver na tela.

Para resolver isso, usamos o método especial `__str__`. Ele ensina ao Python como converter o objeto para uma string amigável quando chamado pelo `print()` ou em uma f-string.

### O método `__str__`

Podemos implementar o método `__str__` para alterar a string que representa o objeto.

```python
class Aluno:
    def __init__(self, nome, matricula, ira):
        self.nome = nome
        self.matricula = matricula
        self.ira = ira

    def __str__(self):
        return f"Aluno: {self.nome} | Matrícula: {self.matricula} | IRA: {self.ira}"
```

Testando a exibição com `print()`

```python
a1 = Aluno("Carla Dias", 3001, 8.7)
print(a1)
```

Nós podemos agrupar os objetos em listas.

```python
# Criando uma lista de objetos
turma = [
    Aluno("Kaio Rodrigues", 3002, 7.5),
    Aluno("Natália Rios", 3003, 9.2),
    Aluno("Renato Souza", 3004, 6.8)
]
```

Dentro do laço, a variável temporária assume cada objeto da lista sequencialmente.

```python
for aluno in turma:
    print(aluno)
```

Podemos acessar atributos numéricos de cada objeto para fazer somatórios, médias ou filtros de busca.

```python
soma_ira = 0

for aluno in turma:
    soma_ira += aluno.ira

media_geral = soma_ira / len(turma)
print(f"IRA médio: {media_geral:.2f}")
```

```python
# Filtrando apenas alunos com IRA igual ou superior a 7.0
print("Alunos aprovados com folga:")

for aluno in turma:
    if aluno.ira >= 7.0:
        print(f"- {aluno.nome} (IRA: {aluno.ira})")
```

### Exercício 3

Crie uma classe chamada `Produto` contendo
* Construtor recebendo `nome`, `preco` e `estoque`.
* Método `__str__` que retorne: `"Produto: <nome> | Preço: R$ <preco> | Estoque: <estoque> un"`.
* Método `valor_total()` que retorne o cálculo `preco * estoque`.

Em seguida
1. Crie uma lista com 3 produtos distintos.
2. Imprima os produtos usando o laço `for`.
3. Some e exiba o patrimônio total em estoque somando os valores totais de cada item da lista.

```python
# Sua solução aqui
class Produto:
    def __init__(self, nome, preco, estoque):
        self.nome = nome
        self.preco = preco
        self.estoque = estoque

    def __str__(self):
        return f'Produto: {self.nome} | Preço: R${self.preco} | Estoque: {self.estoque} un'

    def valor_total(self):
        return self.preco * self.estoque

lista = [Produto('Livro', 50.0, 3),
         Produto('Caneta', 10.0, 5),
         Produto('Marca-Texto', 12.0, 5)]

summ = 0
for p in lista:
    print(p)
    summ += p.valor_total()
print(f'Patrimônio total: R${summ}')
```

## Composição

Sistemas complexos são formados pela união de partes menores:

* Um **computador** é composto por processador, memória e placa de vídeo.
* Uma **playlist** é composta por várias músicas.
* Um **carrinho de compras** é composto por vários produtos.

Em POO, chamamos isso de **composição**: quando uma classe tem como atributo uma lista de instâncias de outra classe.

Primeiro criamos a classe mais simples que representará cada produto.

```python
class Produto:
    def __init__(self, nome, preco):
        self.nome = nome
        self.preco = preco

    def __str__(self):
        return f"{self.nome} - R$ {self.preco:.2f}"
```

O `Carrinho` inicia com uma lista vazia de itens e possui métodos para receber e processar objetos da classe `Produto`.

```python
class Carrinho:
    def __init__(self):
        # O atributo itens começa como uma lista vazia
        self.itens = []

    def adicionar_produto(self, produto):
        self.itens.append(produto)

    def total(self):
        soma = 0
        for item in self.itens:
            soma += item.preco
        return soma

    def exibir_carrinho(self):
        print("~~~~ Carrinho ~~~~")
        print(r'''
 //\
 || \___________
 ||  \_|__|__|__|
  \_ |__|__|__|__|
    \___________/
        O   O
''')
        for item in self.itens:
            print(f"- {item}")
        print(f"Total: R$ {self.total():.2f}")
```

Instanciando os produtos independentes

```python
p1 = Produto("Mouse sem fio", 80.0)
p2 = Produto("Teclado mecânico", 250.0)
p3 = Produto("Mousepad", 45.0)
```

Criando o carrinho e adicionando os objetos criados

```python
meu_carrinho = Carrinho()

meu_carrinho.adicionar_produto(p1)
meu_carrinho.adicionar_produto(p2)
meu_carrinho.adicionar_produto(p3)
```

```python
meu_carrinho.exibir_carrinho()
```

## Atributos de Instância vs. Atributos de Classe

### Dados individuais vs. Dados compartilhados

Até agora, todo atributo que criamos usava `self` (`self.nome`, `self.saldo`). Cada objeto tinha sua própria cópia exclusiva daquele dado.

Porém, às vezes precisamos de uma informação que pertença **à classe como um todo**, sendo compartilhada igualmente por todas as instâncias:

* Um contador para saber quantas instâncias daquela classe já foram criadas no programa.
* Uma constante global da classe (como o nome da instituição ou uma taxa fixa).

**Atributo de Instância:** Pertence ao objeto individual (`self.nome`).

**Atributo de Classe:** Pertence à classe e é compartilhado por todos os objetos dela.

### Definindo um Atributo de Classe

Declaramos o atributo diretamente dentro do bloco da classe, fora de qualquer método (fora do `__init__`).

```python
class Aluno:
    # Atributos de classe (compartilhados por todos os objetos)
    instituicao = "Universidade Federal"
    total_alunos = 0

    def __init__(self, nome):
        # Atributo de instância (único para cada aluno)
        self.nome = nome

        # Incrementa o contador da classe toda vez que um novo aluno nasce
        Aluno.total_alunos += 1
```

Acessamos atributos de classe usando a sintaxe `NomeDaClasse.atributo`.

```python
print(f"Total de alunos no sistema: {Aluno.total_alunos}")
```

Instanciando novos objetos

```python
a1 = Aluno("Eduardo")
a2 = Aluno("Camila")
a3 = Aluno("Rodrigo")
```

Verificando o impacto no atributo de classe compartilhado

```python
print(f"Total de alunos após cadastros: {Aluno.total_alunos}")
```

```python
# Todas as instâncias enxergam a mesma instituição
print(f"{a1.nome} estuda na: {a1.instituicao}")
print(f"{a2.nome} estuda na: {a2.instituicao}")
```

### Especificações do Sistema

#### 1. Classe `Livro`
* **Atributos no construtor:** `titulo`, `autor`, `ano`.
* **Atributo padrão:** `disponivel` (iniciado sempre como `True`).
* **Método `emprestar()`:**
  * Se o livro estiver disponível, altera `disponivel` para `False` e retorna `True`.
  * Se já estiver emprestado, imprime `"Livro indisponível no momento."` e retorna `False`.
* **Método `devolver()`:**
  * Altera `disponivel` para `True`.
* **Método `__str__()`:**
  * Retorna uma string formatada: `"[Disponível] Titulo (Ano) - Autor"` ou `"[Emprestado] Titulo (Ano) - Autor"`.

* **Método `listar_acervo()`:**
  * Imprime todos os livros do acervo utilizando o `__str__` de cada um.
* **Método `buscar_por_autor(autor)`:**
  * Percorre o acervo e imprime todos os livros pertencentes àquele autor.
