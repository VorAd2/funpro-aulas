## 1. Compreensão de listas e o paradigma declarativo

Até o momento, a criação de uma nova lista a partir de uma coleção existente dependia de uma abordagem imperativa: inicializar uma lista vazia, iterar com um laço for e chamar o método append() repetidas vezes.

A compreensão de listas (list comprehension) adota uma abordagem declarativa, inspirada na notação matemática de definição de conjuntos: por exemplo, $\{x^2 \mid x \in \mathbb Z, 1 \leq x < 6\}$.


```python
L = [x ** 2 for x in range(1, 6)]
print(L)
```

```python
# abordagem imperativa tradicional
quadrados_imperativo = []

for numero in range(1, 6):
    quadrados_imperativo.append(numero ** 2)

print("lista gerada de forma imperativa:", quadrados_imperativo)
```

A 'compreensão' de listas permite a inserção de uma cláusula condicional `if` ao final da expressão. O `if` atua como um filtro: apenas os elementos que satisfazem o predicado booleano passam para a etapa de transformação e são incluídos na lista final.

```python
L = [x ** 2 for x in range(1, 6) if x % 2 == 0]
print(L)
```

Quando o `if` está no final da linha, ele define quais itens entram na lista; quando o `if-else` (operador ternário) está no início, ele define qual expressão deve ser aplicada a cada elemento existente.

```python
# transformacao condicional de todos os elementos via operador ternario
notas = [4.5, 8.0, 6.5, 9.0, 3.2]

# projeta uma classificacao sem alterar a quantidade de itens da lista
status_alunos = ["aprovado" if nota >= 7.0 else "recuperacao" for nota in notas]

print("classificacao final:", status_alunos)
```

## 2. Compreensão de dicionários e conjuntos

O mesmo paradigma declarativo aplicado às listas pode ser estendido a outras estruturas de dados, como os dicionários (`dict`) e os conjuntos (`set`).

Na compreensão de dicionários (`dict comprehension`), delimitamos a expressão por chaves `{}` e definimos a regra de formação explicitando um par `chave: valor` antes da cláusula `for`. Esse padrão elimina a necessidade de alocar um dicionário vazio e preenchê-lo por indexação manualmente.

```python
# mapeamento e transformacao de dados em dicionario
produtos = ["teclado", "mouse", "monitor", "headset"]
precos_base = [100.0, 50.0, 800.0, 200.0]

# combinando colecoes e aplicando 10% de imposto sobre os valores
precos_finais = {produtos[i]: precos_base[i] * 1.10 for i in range(len(produtos))}

print("tabela de precos atualizada:", precos_finais)
```

Assim como nas listas, podemos incluir uma cláusula `if` ao final da compreensão para restringir quais pares (chave, valor) devem ser incorporados à coleção resultante.

```python
# filtragem e inversao de chaves e valores em um dicionario
estoque = {"parafuso": 150, "prego": 0, "porca": 85, "arruela": 0}
```

```python
# filtrando apenas itens disponiveis no estoque
disponiveis = {item.upper(): estoque[item] for item in estoque if estoque[item] > 0}
print("itens com saldo positivo:", disponiveis)
```

Um uso clássico é a inversão de papéis entre chaves e valores, iterando sobre o gerador de pares retornado pelo método `items()`.

```python
# inversao de mapeamento de codigos para busca reversa
usuarios_por_id = {101: "ana", 102: "bruno", 103: "ana"}
id_por_usuario = {nome: cod for (cod, nome) in usuarios_por_id.items()}
print("mapeamento invertido:", id_por_usuario)
```

```python
# versão imperativa
usuarios_por_id = {101: "ana", 102: "bruno", 103: "ana"}
D = {}
for (cod, nome) in usuarios_por_id.items():
    D[nome] = cod
print(D)
```

A compreensão de conjuntos (`set comprehension`) adota a mesma sintaxe delimitada por chaves `{}`, mas recebe 1 expressão só, em vez de chave e valor.

```python
# extracao e normalizacao declarativa de valores unicos
cidades_brutas = ["  Fortaleza ", "RECIFE", "fortaleza", " Salvador", "recife", "FORTALEZA"]

# sanitiza espacos, padroniza caixa baixa e elimina duplicatas em uma unica etapa
cidades_unicas = {cidade.strip().lower() for cidade in cidades_brutas}

print("conjunto de cidades unicas e normalizadas:", cidades_unicas)
```

## 3. Compreensões aninhadas e manipulação de matrizes

Uma compreensão pode conter múltiplos laços `for` encadeados. O caso mais comum dessa construção é a linearização (ou flattening), que consiste em converter uma matriz bidimensional (lista de listas) em um vetor unidimensional simples.

A ordem das cláusulas `for` dentro dos colchetes segue exatamente a ordem em que os laços seriam aninhados em blocos tradicionais, com o laço externo declarado primeiro e o laço interno logo em seguida.

```python
# matriz bidimensional representada como lista de listas
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# linearizacao da matriz: primeiro percorre as linhas, depois cada item da linha
vetor_linear = [elemento for linha in matriz for elemento in linha]

print("matriz original:", matriz)
print("vetor linearizado:", vetor_linear)
```

A compreensão aninhada também resolve uma armadilha comum na alocação de matrizes em memória. Ao tentar inicializar uma matriz usando o operador de repetição, como `[[0] * colunas] * linhas`, o Python não copia a lista interna, mas apenas replica ponteiros para a mesma posição na memória física. Qualquer alteração em um índice reflete em todas as outras linhas simultaneamente.

Ao empregar uma compreensão aninhada na forma `[[0 for _ in range(colunas)] for _ in range(linhas)]`, cada passagem do laço externo força a alocação de uma lista completamente nova e isolada na memória RAM.

```python
# demonstracao do erro de compartilhamento de ponteiros
A = [0, 0, 0]
matriz_defeituosa = [A, A, A]
matriz_defeituosa[0][0] = 99
print("matriz com ponteiros duplicados:", matriz_defeituosa)
```

```python
# alocacao correta e independente com compreensao aninhada
# abaixo, usar A em vez de [0 for j in range(3)] daria o mesmo problema de antes...
A = [0, 0, 0]
matriz_independente = [[0 for j in range(3)] for i in range(3)]
matriz_independente[0][0] = 99
print("matriz com linhas independentes:", matriz_independente)

# escrever [0, 0, 0] gera uma nova lista
# do mesmo jeito, escrever list comprehension [x for ...] gera uma nova lista
# [[0, 0, 0] for ...] está ok, pois a lista interna é nova!
# [[x for ...] for ...] está ok, pois a lista interna é nova!
# [A for ...] não é ok, pois a lista interna (A) é compartilhada
# [[0] * N for ...] está ok, pois a lista interna é nova: [0] * N é [0, ..., 0];
# apenas o zero é repetido: isso não tem problema, pois int é imutável
# [[0, 0, 0]] * M não está ok, pois a repetição (*) compartilha a lista interna
# [A] * M não está ok, pois [A] * M é [A, A, ..., A]: o A é compartilhado
```

**Veja acima** que o problema não é `for` vs `*`, é sobre uma lista ser copiada ou criada nova. Teste cada um dos casos, para ajudar a entender! Altere a célula acima.

Podemos aplicar a mesma lógica para expressar transformações lineares, como a transposição de uma matriz, que consiste em transformar linhas em colunas.

Para transpor uma matriz de dimensões $M \times N$, invertemos a ordem dos índices, iterando pelas colunas no laço externo e pelas linhas no laço interno.

```python
M = [
    [10, 20, 30],
    [40, 50, 60]
]
T = [[M[i][j] for i in (0, 1)] for j in (0, 1, 2)]

print(T)
```

```python
M = [
    [10, 20, 30],
    [40, 50, 60]
]
T = []
for j in (0, 1, 2):
    L = []
    for i in (0, 1):
        L.append(M[i][j])
    T.append(L)
```

```python
# matriz retangular 2x3
matriz_original = [
    [10, 20, 30],
    [40, 50, 60]
]

total_linhas = len(matriz_original)
total_colunas = len(matriz_original[0])

# transposicao declarativa para formato 3x2
matriz_transposta = [[matriz_original[i][j] for i in range(total_linhas)] for j in range(total_colunas)]

print("matriz original (2x3):", matriz_original)
print("matriz transposta (3x2):", matriz_transposta)
```

```python
# extracao e filtragem de elementos de uma matriz com condicional
# seleciona apenas valores impares presentes em qualquer linha da matriz
impares_matriz = [item for linha in matriz for item in linha if item % 2 != 0]

print("elementos impares extraidos:", impares_matriz)
```

## 4. Funções anônimas e a palavra-chave lambda

O conceito de função lambda tem origem no cálculo lambda ($\lambda$-cálculo), um formalismo matemático criado por [Alonzo Church](https://pt.wikipedia.org/wiki/Alonzo_Church) na década de 1930 para investigar computabilidade e abstração funcional antes mesmo da invenção dos computadores eletrônicos.

Em linguagens modernas, funções são cidadãs de primeira classe (first-class citizens), o que significa que podem ser instanciadas, armazenadas em variáveis e repassadas como parâmetros.

A palavra-chave `def` cria uma função nomeada, abrindo um bloco de código que aceita múltiplas instruções e requer a cláusula `return` explícita para devolver um valor. Em contraste, a palavra-chave `lambda` constrói uma função anônima inline (em uma única linha). A sintaxe aceita zero ou mais parâmetros e calcula uma única expressão, cujo resultado é retornado de forma implícita.

Uma função `lambda` não aceita comandos complexos como laços `while`, blocos `try-except` ou atribuições de variáveis em seu corpo; ela avalia uma única expressão.

```python
# comparacao entre funcao tradicional e expressao lambda equivalente
def quadrado_nomeado(x):
    return x ** 2

# definicao da mesma operacao via expressao lambda
quadrado_lambda = lambda x: x ** 2
# claro, acima você guardou a função anônima em uma variável
# ou seja, acabou dando um nome... até que alguém guarde outro objeto na variável

print("chamada com def:", quadrado_nomeado(5))
print("chamada com lambda:", quadrado_lambda(5))
```

```python
# funcao lambda com multiplos parametros
calcula_desconto = lambda valor, percentual: round(valor * (1 - percentual / 100), 2)

preco_final = calcula_desconto(150.0, 15)
print("preco calculado com desconto:", preco_final)
```

O uso de `lambda` associado diretamente a uma variável é incomum, pois uma função regular declarada com `def` oferece mensagens de rastreamento de erro (traceback) mais claras durante a depuração.

O emprego ideal de funções `lambda` ocorre quando precisamos fornecer uma regra pontual para uma função de ordem superior (uma função que recebe outra função como parâmetro).

O cenário mais clássico é o uso do argumento `key` nas funções `sorted()`, `min()` e `max()`, onde a função `lambda` dita qual critério numérico ou textual deve guiar o algoritmo de ordenação sobre estruturas complexas como tuplas ou dicionários.

```python
# ordenacao personalizada de lista de dicionarios usando o parametro key
alunos = [
    {"nome": "carlos", "nota": 7.5, "idade": 21},
    {"nome": "bia", "nota": 9.2, "idade": 19},
    {"nome": "alice", "nota": 6.8, "idade": 24},
    {"nome": "daniel", "nota": 8.1, "idade": 20},
    {"nome": "jorge", "nota": 7.5, "idade": 20},
    {"nome": "joana", "nota": 7.5, "idade": 22}
]
def criterio(d):
    return d["nota"]

# ordenando os alunos pela nota em ordem decrescente
# poderia passar key=criterio tambeḿ
alunos_por_nota = sorted(alunos, key=lambda d: d["nota"], reverse=True)

print("alunos ordenados por nota decrescente:")
for registro in alunos_por_nota:
    print(registro["nome"], "->", registro["nota"], " | idade:", registro["idade"])
```

```python
# ordenando os alunos pela nota em ordem decrescente, idade para desempate
alunos_por_nota_e_idade = sorted(alunos, key=lambda aluno: (aluno["nota"], aluno["idade"]), reverse=True)

print("alunos ordenados por nota decrescente:")
for registro in alunos_por_nota_e_idade:
    print(registro["nome"], "->", registro["nota"], " | idade:", registro["idade"])
```

```python
# localizando registros extremos com min e max guiados por chave funcional
aluno_mais_novo = min(alunos, key=lambda d: d["idade"])
aluno_maior_nota = max(alunos, key=lambda aluno: aluno["nota"])

print("aluno com menor idade:", aluno_mais_novo["nome"], f"({aluno_mais_novo['idade']} anos)")
print("aluno com maior nota:", aluno_maior_nota["nome"], f"({aluno_maior_nota['nota']})")
```

## 5. Funções de ordem superior

Uma função de ordem superior (higher-order function) é qualquer função que aceita outras funções como argumento ou retorna uma função como resultado.

Como as funções em Python são objetos comuns, o identificador de uma função armazena apenas uma referência para o bloco de código compilado na memória. Quando passamos esse identificador como parâmetro para outra função, estamos repassando a capacidade de invocar esse código através do operador de chamada `()`.

Esse mecanismo permite desacoplar a estrutura do algoritmo (como percorrer uma coleção ou realizar um cálculo numérico) da regra específica que deve ser aplicada a cada elemento.

```python
# definindo uma funcao de ordem superior que transforma coleções
def aplicar_transformacao(colecao, operacao):
    resultado = []
    for item in colecao:
        resultado.append(operacao(item))
    return resultado
```

```python
valores = [1, 2, 3, 4, 5]
```

```python
L = aplicar_transformacao(valores, lambda x: x ** 2)
print(L)
```

```python
# passando outra funcao lambda para converter graus celsius em fahrenheit
celsius = [0, 20, 37, 100]
fahrenheit = aplicar_transformacao(celsius, lambda c: (c * 9/5) + 32)
print("temperaturas convertidas:", fahrenheit)
```

O mesmo princípio se aplica à filtragem por predicados. Um predicado é uma função que avalia um dado e retorna um valor booleano (`True` ou `False`).

Ao construir uma função genérica de filtragem que recebe um predicado como argumento, eliminamos a necessidade de duplicar laços `for` e estruturas condicionais para cada novo critério de busca.

```python
# definindo uma funcao de ordem superior para selecao condicional
def filtrar_colecao(colecao, predicado):
    selecionados = []
    for item in colecao:
        if predicado(item):
            selecionados.append(item)
    return selecionados
```

```python
palavras = ["algoritmo", "cpu", "memoria", "io", "processador", "ram"]
```

```python
# filtrando apenas termos com mais de 5 caracteres
termos_longos = filtrar_colecao(palavras, lambda termo: len(termo) > 5)
print("termos longos:", termos_longos)
```

```python
# filtrando termos que contem a letra 'o'
termos_com_o = filtrar_colecao(palavras, lambda termo: "o" in termo)
print("termos que contem a letra o:", termos_com_o)
```

Em computação científica e métodos numéricos, funções de ordem superior são a base para algoritmos genéricos.

Podemos definir uma função para calcular a derivada numérica de qualquer função matemática unidimensional $f(x)$ em um ponto $x_0$, empregando a definição clássica do limite por diferenças finitas:

$$\frac{f(x_0 + h) - f(x_0)}{h}$$

O algoritmo de derivação não precisa conhecer previamente a fórmula da curva; ele recebe a função matemática $f$ arbitrariamente por parâmetro.

```python
# calculo generico de derivada numerica via diferencas finitas
def derivada_numerica(funcao_matematica, ponto_x, h=1e-5):
    inclinacao = (funcao_matematica(ponto_x + h) - funcao_matematica(ponto_x)) / h
    return round(inclinacao, 4)
```

```python
# calculando a derivada de f(x) = x^2 no ponto x = 3 (valor exato teorico: 2*3 = 6)
derivada_parabola = derivada_numerica(lambda x: x ** 2, 3)
print("derivada de x^2 em x=3:", derivada_parabola)
```

```python
# calculando a derivada de f(x) = 2*x^3 - 5*x no ponto x = 2 (valor exato teorico: 6*(2^2) - 5 = 19)
derivada_cubica = derivada_numerica(lambda x: 2 * (x ** 3) - 5 * x, 2)
print("derivada de 2x^3 - 5x em x=2:", derivada_cubica)
```

Onde aceita função anônima também aceita função nomeada:

```python
def dobro(x):
    return x * 2

def cubo(x):
    return x ** 3

print("aplicando funcao nomeada dobro:", aplicar_transformacao([1, 2, 3, 4], dobro))
print("aplicando funcao nomeada cubo:", aplicar_transformacao([1, 2, 3, 4], cubo))
```

```python
n = 4 # deve funcionar para qualquer 'n'

I = [[1 if i == j else 0 for j in range(n)] for i in range(n)]

for l in I:
    print(l)
print()

D = [I[i][j] for i in range(n) for j in range(n) if i == j]
print(D)
```

### Exercício 3: Acumulador de ordem superior

Algoritmos de redução (como somatórios, produtórios e agregações) percorrem uma coleção acumulando um resultado a cada iteração. Implemente sua própria função de ordem superior `reduzir_colecao(colecao, operacao, valor_inicial)` que recebe uma coleção, uma função de agregação de dois argumentos e um valor inicial de acumulação.

Em seguida, aplique essa função para totalizar o faturamento de pedidos e utilize a função embutida `sorted()` com o parâmetro `key` configurado via `lambda` para realizar uma ordenação multicritério sobre os pedidos: primeiro pela prioridade da entrega e, em caso de empate, pelo valor do pedido em ordem decrescente.

```python
# DEFINA reduzir_colecao AQUI
def reduzir_colecao(colecao, operador, valor_inicial):
    total = 0
    for pedido in colecao:
        total = operador(total, pedido)
    return total

pedidos = [
    {"id": 1, "prioridade": "baixa", "valor": 250.0},
    {"id": 2, "prioridade": "alta", "valor": 1200.0},
    {"id": 3, "prioridade": "media", "valor": 450.0},
    {"id": 4, "prioridade": "alta", "valor": 850.0},
    {"id": 5, "prioridade": "media", "valor": 980.0}
]

total_pedidos = reduzir_colecao(pedidos, lambda acumulado, pedido: acumulado + pedido["valor"], 0)
print("faturamento total acumulado:", round(total_pedidos, 2))
```

```python
# COMPLETE ESTA CÉLULA (dica: mapeie as strings 'alta', 'media', 'baixa' para inteiros apropriados; pode usar um dict)
D = {
    'alta': 1,
    'media': 2,
    'baixa': 3
}
pedidos_ordenados = sorted(pedidos, key=lambda p: (D[p['prioridade']], -p['valor']))

print("pedidos organizados por prioridade e valor decrescente:")
for registro in pedidos_ordenados:
    print(f"prioridade: {registro['prioridade']} | valor: {registro['valor']} | id: {registro['id']}")
```
